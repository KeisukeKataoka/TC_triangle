```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
import seaborn as sns
import fssa

# システムサイズ
L_list = np.array([8, 16, 32, 64])

# 温度パラメータ
T = np.linspace(2.0, 2.8, 50)

# 臨界温度 (2D Ising)
Tc_exact = 2.269

# フェイクの物理量（例えば磁化）
# finite-size scaling: M ~ L^{-beta/nu} f((T-Tc) L^{1/nu})
beta = 0.125
nu = 1.0

M = np.array([ (1.0 - (T/Tc_exact)**2)**beta * L**(-beta/nu) for L in L_list ])
dM = 0.02 * np.ones_like(M)  # 仮の誤差
```

```python
import itertools
from cycler import cycler
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import fssa
################
L_list=[16,20,24,28] #,36]
#a=np.zeros((len(L_list),18),dtype='uint32')
########
#################
data_set = {}
p_list=[]
L_list=[10,12,16,20,22,24]
#L2_list=[16**2,20**2,24**2,28**2]
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
#fig, ax = plt.subplots(1,2,figsize=(12,4))
Np=31
d=4
gamma=0.1
#fig, ax = plt.subplots(1,2,figsize=(12,4))
########
########
cmap = cm.get_cmap('gnuplot')
########
########
for ip in range(Np):
    pdp=0.5*ip/(Np-1)+0.25
    p_list=np.append(p_list,pdp)
a=[]
da=[]
#fig, ax = plt.subplots(1,2,figsize=(12,4))
for tmp_i in range(7):
    Lx=L_list[tmp_i]
    Ly=L_list[tmp_i]
    data_set= np.load("Lx="+"{:}".format(Lx)+"ZX-ver1.npz")
    #print(Lx,Ly)
    R2csr_ave=[]
    R2csr_error=[]
    R2csr_var=[]
    for ip in range(Np):
        # stabilizer matrix
        #pdp=ip/(Np-1)
        pdp=0.5*ip/(Np-1)+0.25
        key1="Lx="+"{:}".format(Lx)+"_Ly="+"{:}".format(Ly)+"_p="+"{:.2f}".format(pdp)
        #print(data_set)
        R2csr_list=data_set[key1+"zx"]*Ly*(Lx-3)
        ###
        #print(R2_csrlist)
        R2csr_aved=np.mean(R2csr_list)
        R2csr_error_d=np.std(R2csr_list, ddof=1)/ np.sqrt(len(R2csr_list))
        R2csr_vard=np.var(R2csr_list)/Ly/(Lx-3)
        ###
        R2csr_ave=np.append(R2csr_ave,R2csr_aved)
        R2csr_error=np.append(R2csr_error,R2csr_error_d)
        R2csr_var=np.append(R2csr_var,R2csr_vard)
    v1s=R2csr_var[d:-d]
    a+=[v1s.tolist()]
    vd1=v1s*gamma
    da+=[vd1.tolist()]
pc_ini= [0.5] #np.linspace(0.70,0.8,10)
#print(a)
S_list=[]
for ip in pc_ini:
    print(ip)
    #p_list_r=p_list[3:18]
    ##############
    bounds = -1,1
    #ret = fssa.autoscale(L_list, p_list, a, da, 0.4, 1.8, 2.2)
    ret = fssa.autoscale(L_list, p_list[d:-d], a,da,ip,2,2,x_bounds=None)
    auto_scaled_data = fssa.scaledata(L_list, p_list[d:-d], a, da, ret.rho, ret.nu, ret.zeta)
########
    fig, ax = plt.subplots(figsize=(4,4))
    ax.tick_params(labelsize=14,colors = "blue")
    #ax.set_prop_cycle(cycler('color', palette))
    for tmp_i in range(7):
        ax.scatter(auto_scaled_data.x.T[:,tmp_i], auto_scaled_data.y.T[:,tmp_i], marker=markers[tmp_i],color=cmap((tmp_i+1)/8),s=80)
    #ax.set_xbound(-4, 2)
    #plt.legend()
    #ax.set_xlabel(r'$x$')
    #plt.savefig('var_chi_scaling_func.svg') # 画像の保存
    ax.tick_params(axis='x', colors='blue')  # x軸目盛り
    ax.tick_params(axis='y', colors='blue')  # y軸目盛り
    ax.tick_params(axis='both', which='major', labelsize=22)
    ax.set_xlabel(r'$(r-r_c)L_t^{\frac{1}{\nu}}$',fontsize=24)  # x軸ラベル
    plt.ylim(-0.005,0.06)
    ax.set_ylabel(r'$F$',fontsize=24)  # y軸ラベル
    plt.savefig('fitt.svg',dpi=300) # 画像の保存
    plt.show()
#######
    # critical exponents and errors, quality of data collapse
    print("pc",ret.rho, ret.drho)
    print("nu",ret.nu, ret.dnu)
    print("zeta",ret.zeta, ret.dzeta,)
    #print(ret.fun)

    ##scaled data##
    #print(auto_scaled_data)
    xs,ys,dys = fssa.scaledata(L_list, p_list[d:-d], a, da, ret.rho, ret.nu, ret.zeta)

    ##quality##
    S=fssa.quality(xs, ys, dys, x_bounds=None)
    print("S",S)
    #print("quality",S)
    S_list=np.append(S_list,S)
```
