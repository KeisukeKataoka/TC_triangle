```python
import numpy as np
import glob
import re

files = glob.glob("Lx_*_Nd_100_NT_3456_px_0.0_MPI_hexagonal1234.npz")

Ls = []
pg_list = []

combined = {key: [] for key in [
    'TEE1_ave','TEE1_err','TEE1_var',
    'TEE2_ave','TEE2_err','TEE2_var',
    'TEE3_ave','TEE3_err','TEE3_var',
    'TEE4_ave','TEE4_err','TEE4_var',
]}

for filename in files:
    match = re.search(r"Lx_(\d+)_", filename)
    if not match:
        continue
    Lx = int(match.group(1))
    Ls.append(Lx)

    data = np.load(filename)

    pg_list.append(data['pg'])

    for key in combined.keys():
        combined[key].append(data[key])

for key in combined:
    combined[key] = np.array(combined[key])  # shape = (num_Lx, len(pg))

pg_all = np.array(pg_list)  # shape = (num_Lx, len(pg))

# Ls昇順にソート
sort_idx = np.argsort(Ls)
Ls = np.array(Ls)[sort_idx]
pg_all = pg_all[sort_idx]
for key in combined:
    combined[key] = combined[key][sort_idx]

# 保存
np.savez("test.npz", Ls=Ls, pg=pg_all, **combined)
```
