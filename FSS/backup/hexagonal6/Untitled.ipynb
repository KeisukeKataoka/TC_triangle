```python
import numpy as np

# 測定確率のリスト
pg_list = np.linspace(0, 1, 21)  # 横軸
px_list = np.linspace(0, 1, 21)  # 縦軸

# 各 (px, pg) に対する秩序指標（TEEなど）
# ここではダミーデータとしてランダム値
TEE_data = np.random.rand(len(px_list), len(pg_list))  # shape=(len(px_list), len(pg_list))

import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))

# pg（横）と px（縦）のメッシュグリッド
PG, PX = np.meshgrid(pg_list, px_list)

# ヒートマップ描画
c = plt.pcolormesh(PG, PX, TEE_data, shading='auto', cmap='viridis')
plt.colorbar(c, label='TEE (topological order)')

# 軸ラベル
plt.xlabel('Measurement probability $p_g$')
plt.ylabel('Measurement probability $p_x$')
plt.title('Phase diagram (TEE)')

plt.show()
```
