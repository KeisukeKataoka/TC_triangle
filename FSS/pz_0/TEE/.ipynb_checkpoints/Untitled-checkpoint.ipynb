```python
import numpy as np
import glob
import fssa
import matplotlib.pyplot as plt
from cycler import cycler
import matplotlib.cm as cm

# --------------------------
# 設定
# --------------------------
d = 0           # 端データ除外（必要に応じて調整）
gamma = 0.5     # 誤差スケーリング
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
cmap = cm.get_cmap('gnuplot')
pc_ini = [0.5,0.6,0,7,0.8]  # 初期臨界点

# --------------------------
# データ読み込み
# --------------------------
files = sorted(glob.glob("Lx_*_*.npz"))
L_list = []
TEE_data = {}

for f in files:
    # ファイル名から Lx を取得
    Lx = int(f.split("_")[1])
    L_list.append(Lx)
    
    # npz からデータ取得
    data_set = np.load(f)
    p_list = data_set['pg']

    for tee_idx in range(1,5):
        a = data_set[f'TEE{tee_idx}_var']   # 分散
        da = data_set[f'TEE{tee_idx}_err'] * gamma  # 誤差 × γ

        # 端データ除外
        if d > 0:
            a = a[d:-d]
            da = da[d:-d]
            p_use = p_list[d:-d]
        else:
            p_use = p_list

        key = f'TEE{tee_idx}'
        if key not in TEE_data:
            TEE_data[key] = {'a': [], 'da': [], 'p': p_use}  # p は同じなので1回だけでOK
        TEE_data[key]['a'].append(a.tolist())
        TEE_data[key]['da'].append(da.tolist())

# --------------------------
# FSS 実行
# --------------------------
for tee_idx in range(1,5):
    a = TEE_data[f'TEE{tee_idx}']['a']
    da = TEE_data[f'TEE{tee_idx}']['da']
    p_use = TEE_data[f'TEE{tee_idx}']['p']

    for ipc in pc_ini:
        # autoscaleで臨界点・指数を最適化
        ret = fssa.autoscale(L_list, p_use, a, da, ipc, 2, 2, x_bounds=None)
        auto_scaled_data = fssa.scaledata(L_list, p_use, a, da, ret.rho, ret.nu, ret.zeta)

        # プロット
        fig, ax = plt.subplots(figsize=(4,4))
        for i, L in enumerate(L_list):
            ax.scatter(auto_scaled_data.x.T[:,i], auto_scaled_data.y.T[:,i],
                       marker=markers[i % len(markers)],
                       color=cmap((i+1)/8), s=80)
        ax.set_xlabel(r'$(r-r_c)L^{1/\nu}$', fontsize=18)
        ax.set_ylabel(r'$F$', fontsize=18)
        ax.tick_params(axis='both', which='major', labelsize=14)
        plt.title(f'TEE{tee_idx}')
        plt.show()

        # 結果出力
        print(f"TEE{tee_idx}  pc:", ret.rho, "±", ret.drho)
        print(f"TEE{tee_idx}  nu:", ret.nu, "±", ret.dnu)
        print(f"TEE{tee_idx}  zeta:", ret.zeta, "±", ret.dzeta)

        # スケーリング品質
        xs, ys, dys = fssa.scaledata(L_list, p_use, a, da, ret.rho, ret.nu, ret.zeta)
        S = fssa.quality(xs, ys, dys, x_bounds=None)
        print(f"TEE{tee_idx} S (quality):", S)
```

```python
import numpy as np
import glob
import fssa
import matplotlib.pyplot as plt
from cycler import cycler
import matplotlib.cm as cm

# --------------------------
# 設定
# --------------------------
d = 0               # 端データ除外
gamma = 0.2         # 誤差スケーリング
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
cmap = cm.get_cmap('gnuplot')
pc_ini = [0.75, 0.76, 0.77]  # 初期臨界点候補

# --------------------------
# データ読み込み
# --------------------------
files = sorted(glob.glob("Lx_*_*.npz"))
L_list = []
TEE_data = {}

for f in files:
    Lx = int(f.split("_")[1])
    if Lx < 4:  # 小さい系を除外
        continue
    L_list.append(Lx)

    data_set = np.load(f)
    p_list = data_set['pg']

    for tee_idx in range(1,5):
        a = data_set[f'TEE{tee_idx}_var']
        da = data_set[f'TEE{tee_idx}_err'] * gamma

        # 端データ除外
        if d > 0:
            a = a[d:-d]
            da = da[d:-d]
            p_use = p_list[d:-d]
        else:
            p_use = p_list

        # 臨界点周辺に絞る（必要なら）
        pc_guess = 0.7
        idx_min = max(0, np.searchsorted(p_use, pc_guess - 0.05))
        idx_max = min(len(p_use), np.searchsorted(p_use, pc_guess + 0.05))
        a = a[idx_min:idx_max]
        da = da[idx_min:idx_max]
        p_use_cut = p_use[idx_min:idx_max]

        # L ごとに標準化（平均0、標準偏差1）して autoscale に渡す
        a_norm = (a - np.mean(a)) / np.std(a)
        da_norm = (da / np.std(a))  # 誤差も同じ基準でスケーリング

        key = f'TEE{tee_idx}'
        if key not in TEE_data:
            TEE_data[key] = {'a': [], 'da': [], 'p': p_use_cut}
        TEE_data[key]['a'].append(a_norm)
        TEE_data[key]['da'].append(da_norm)

# --------------------------
# FSS 実行
# --------------------------
for tee_idx in range(1,5):
    a = TEE_data[f'TEE{tee_idx}']['a']
    da = TEE_data[f'TEE{tee_idx}']['da']
    p_use = TEE_data[f'TEE{tee_idx}']['p']

    best_S = np.inf
    best_ret = None

    for ipc in pc_ini:
        try:
            ret = fssa.autoscale(L_list, p_use, a, da, ipc, 2, 2, x_bounds=None)
            xs, ys, dys = fssa.scaledata(L_list, p_use, a, da, ret.rho, ret.nu, ret.zeta)
            S = fssa.quality(xs, ys, dys, x_bounds=None)

            if S < best_S:
                best_S = S
                best_ret = ret
        except Exception as e:
            print(f"pc_ini={ipc} でエラー: {e}")

    if best_ret is None:
        print(f"TEE{tee_idx} は全ての初期値で失敗")
        continue

    # プロット
    auto_scaled_data = fssa.scaledata(L_list, p_use, a, da, best_ret.rho, best_ret.nu, best_ret.zeta)
    fig, ax = plt.subplots(figsize=(4,4))
    for i, L in enumerate(L_list):
        ax.scatter(auto_scaled_data.x.T[:,i], auto_scaled_data.y.T[:,i],
                   marker=markers[i % len(markers)],
                   color=cmap((i+1)/8), s=80)
    ax.set_xlabel(r'$(r-r_c)L^{1/\nu}$', fontsize=18)
    ax.set_ylabel(r'$F$', fontsize=18)
    ax.tick_params(axis='both', which='major', labelsize=14)
    plt.title(f'TEE{tee_idx}')
    plt.show()

    # 結果出力
    print(f"TEE{tee_idx}  pc:", best_ret.rho, "±", best_ret.drho)
    print(f"TEE{tee_idx}  nu:", best_ret.nu, "±", best_ret.dnu)
    print(f"TEE{tee_idx}  zeta:", best_ret.zeta, "±", best_ret.dzeta)
    print(f"TEE{tee_idx} S (quality):", best_S)
```

```python
import numpy as np
import glob
import fssa
import matplotlib.pyplot as plt
from cycler import cycler
import matplotlib.cm as cm

# --------------------------
# 設定
# --------------------------
d = 0               # 端データ除外
gamma = 0.2         # 誤差スケーリング
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
cmap = cm.get_cmap('gnuplot')
pc_ini = [0.80, 0.81, 0.82]  # 初期臨界点候補

# --------------------------
# データ読み込み
# --------------------------
files = sorted(glob.glob("Lx_*_*.npz"))
L_list = []
TEE_data = {}  # {'mean':{}, 'var':{}}

for f in files:
    Lx = int(f.split("_")[1])
    if Lx < 4:
        continue
    L_list.append(Lx)

    data_set = np.load(f)
    p_list = data_set['pg']

    for mode in ["mean", "var"]:
        key_mode = "mean" if mode=="mean" else "var"
        if key_mode not in TEE_data:
            TEE_data[key_mode] = {}

        for tee_idx in range(1,5):
            a = data_set[f'TEE{tee_idx}_ave'] if mode=="mean" else data_set[f'TEE{tee_idx}_var']
            da = data_set[f'TEE{tee_idx}_err'] * gamma

            # 端データ除外
            if d > 0:
                a = a[d:-d]
                da = da[d:-d]
                p_use = p_list[d:-d]
            else:
                p_use = p_list

            # データ格納
            key_tee = f'TEE{tee_idx}'
            if key_tee not in TEE_data[key_mode]:
                TEE_data[key_mode][key_tee] = {'a': [], 'da': [], 'p': p_use}
            TEE_data[key_mode][key_tee]['a'].append(a.tolist())
            TEE_data[key_mode][key_tee]['da'].append(da.tolist())

# --------------------------
# FSS 実行
# --------------------------
for mode in ["mean", "var"]:
    print(f"\n===== Mode: {mode} =====")
    for tee_idx in range(1,5):
        data = TEE_data[mode][f'TEE{tee_idx}']
        a = data['a']
        da = data['da']
        p_use = data['p']

        best_S = np.inf
        best_ret = None

        for ipc in pc_ini:
            try:
                ret = fssa.autoscale(L_list, p_use, a, da, ipc, 2, 2, x_bounds=None)
                xs, ys, dys = fssa.scaledata(L_list, p_use, a, da, ret.rho, ret.nu, ret.zeta)
                S = fssa.quality(xs, ys, dys, x_bounds=None)
                if S < best_S:
                    best_S = S
                    best_ret = ret
            except Exception as e:
                print(f"pc_ini={ipc} でエラー: {e}")

        if best_ret is None:
            print(f"TEE{tee_idx} は全ての初期値で失敗")
            continue

        # プロット
        auto_scaled_data = fssa.scaledata(L_list, p_use, a, da, best_ret.rho, best_ret.nu, best_ret.zeta)
        fig, ax = plt.subplots(figsize=(4,4))
        for i, L in enumerate(L_list):
            ax.scatter(auto_scaled_data.x.T[:,i], auto_scaled_data.y.T[:,i],
                       marker=markers[i % len(markers)],
                       color=cmap((i+1)/8), s=80)
        ax.set_xlabel(r'$(r-r_c)L^{1/\nu}$', fontsize=18)
        ax.set_ylabel(r'$F$', fontsize=18)
        ax.tick_params(axis='both', which='major', labelsize=14)
        plt.title(f'TEE{tee_idx} ({mode})')
        plt.show()

        # 結果出力
        print(f"TEE{tee_idx} pc: {best_ret.rho} ± {best_ret.drho}")
        print(f"TEE{tee_idx} nu: {best_ret.nu} ± {best_ret.dnu}")
        print(f"TEE{tee_idx} zeta: {best_ret.zeta} ± {best_ret.dzeta}")
        print(f"TEE{tee_idx} S (quality): {best_S}")
```

```python
import numpy as np
import glob
import fssa
import matplotlib.pyplot as plt
from cycler import cycler
import matplotlib.cm as cm

d = 0
gamma = 0.2
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
cmap = cm.get_cmap('gnuplot')
pc_ini = [0.75, 0.76, 0.77]

files = sorted(glob.glob("Lx_*_*.npz"))
L_list = []
TEE_data = {}

for f in files:
    Lx = int(f.split("_")[1])
    L_list.append(Lx)
    data_set = np.load(f)
    p_list = data_set['pg']

    for tee_idx in range(1,5):
        a = data_set[f'TEE{tee_idx}_var']
        da = data_set[f'TEE{tee_idx}_err'] * gamma

        if d > 0:
            a = a[d:-d]
            da = da[d:-d]
            p_use = p_list[d:-d]
        else:
            p_use = p_list

        # 転移付近の重み付け
        pc_guess = 0.75
        da_weighted = da + 0.01 * np.abs(p_use - pc_guess)

        key = f'TEE{tee_idx}'
        if key not in TEE_data:
            TEE_data[key] = {'a': [], 'da': [], 'p': p_use}
        TEE_data[key]['a'].append(a.tolist())
        TEE_data[key]['da'].append(da_weighted.tolist())

# FSS 実行
for tee_idx in range(1,5):
    a = TEE_data[f'TEE{tee_idx}']['a']
    da = TEE_data[f'TEE{tee_idx}']['da']
    p_use = TEE_data[f'TEE{tee_idx}']['p']

    best_S = np.inf
    best_ret = None

    for ipc in pc_ini:
        try:
            ret = fssa.autoscale(L_list, p_use, a, da, ipc, 2, 2, x_bounds=None)
            xs, ys, dys = fssa.scaledata(L_list, p_use, a, da, ret.rho, ret.nu, ret.zeta)
            S = fssa.quality(xs, ys, dys)
            if S < best_S:
                best_S = S
                best_ret = ret
        except Exception as e:
            print(f"pc_ini={ipc} でエラー: {e}")

    if best_ret is None:
        print(f"TEE{tee_idx} は全ての初期値で失敗")
        continue

    auto_scaled_data = fssa.scaledata(L_list, p_use, a, da, best_ret.rho, best_ret.nu, best_ret.zeta)
    fig, ax = plt.subplots(figsize=(4,4))
    for i, L in enumerate(L_list):
        ax.scatter(auto_scaled_data.x.T[:,i], auto_scaled_data.y.T[:,i],
                   marker=markers[i % len(markers)],
                   color=cmap((i+1)/8), s=80)
    ax.set_xlabel(r'$(p-p_c)L^{1/\nu}$', fontsize=18)
    ax.set_ylabel(r'$F$', fontsize=18)
    plt.title(f'TEE{tee_idx}')
    plt.show()

    print(f"TEE{tee_idx} pc:", best_ret.rho, "±", best_ret.drho)
    print(f"TEE{tee_idx} nu:", best_ret.nu, "±", best_ret.dnu)
    print(f"TEE{tee_idx} zeta:", best_ret.zeta, "±", best_ret.dzeta)
    print(f"TEE{tee_idx} S (quality):", best_S)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import fssa
from cycler import cycler
import matplotlib.cm as cm

# --------------------------
# 設定
# --------------------------
L_list = [10, 12, 16, 20]  # 系の大きさ
p_list = np.linspace(0.6, 0.9, 21)  # 制御パラメータ
pc_true = 0.75   # 臨界点
nu_true = 1.0    # 臨界指数
zeta_true = 0.0  # Fの形状パラメータ
gamma = 0.2      # 誤差スケーリング
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
cmap = cm.get_cmap('gnuplot')

# --------------------------
# データ作成（スケーリングするF）
# --------------------------
TEE_data = {'a': [], 'da': [], 'p': p_list}

for L in L_list:
    x = (p_list - pc_true) * L**(1/nu_true)
    # 仮のスケーリング関数（シグモイド風）
    F = 0.5 * (1 + np.tanh(x))
    # 分散データとして少しノイズを加える
    a = (F * (1-F)) + 0.02*np.random.rand(len(F))
    da = gamma * np.sqrt(a / L)  # 誤差も少し小さくする
    TEE_data['a'].append(a.tolist())
    TEE_data['da'].append(da.tolist())

# --------------------------
# FSS 実行
# --------------------------
rho_c0 = 0.7  # 臨界点初期値
nu0 = 0.5     # 初期指数
zeta0 = 0.5   # 初期 zeta

best_ret = fssa.autoscale(L_list, p_list, TEE_data['a'], TEE_data['da'],
                          rho_c0, nu0, zeta0, x_bounds=None)

auto_scaled_data = fssa.scaledata(L_list, p_list, TEE_data['a'], TEE_data['da'],
                                  best_ret.rho, best_ret.nu, best_ret.zeta)

# --------------------------
# プロット
# --------------------------
fig, ax = plt.subplots(figsize=(5,5))
for i, L in enumerate(L_list):
    ax.scatter(auto_scaled_data.x.T[:,i], auto_scaled_data.y.T[:,i],
               marker=markers[i % len(markers)],
               color=cmap((i+1)/8), s=80)
ax.set_xlabel(r'$(p-p_c)L^{1/\nu}$', fontsize=18)
ax.set_ylabel(r'$F$', fontsize=18)
ax.tick_params(axis='both', which='major', labelsize=14)
plt.title('Sample FSS collapse')
plt.show()

# --------------------------
# 結果出力
# --------------------------
print("pc:", best_ret.rho, "±", best_ret.drho)
print("nu:", best_ret.nu, "±", best_ret.dnu)
print("zeta:", best_ret.zeta, "±", best_ret.dzeta)
S = fssa.quality(auto_scaled_data.x, auto_scaled_data.y, auto_scaled_data.dy)
print("S (quality):", S)
```

```python
import numpy as np
import glob
import fssa
import matplotlib.pyplot as plt
from cycler import cycler
import matplotlib.cm as cm

# --------------------------
# 設定
# --------------------------
d = 0               # 端データ除外
gamma = 0.2         # 誤差スケーリング
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
cmap = cm.get_cmap('gnuplot')
pc_ini = [0.80, 0.81, 0.82]  # 初期臨界点候補

# --------------------------
# データ読み込み
# --------------------------
files = sorted(glob.glob("Lx_*_*.npz"))
L_list = []
TEE_data = {}  # {'mean':{}, 'var':{}}

for f in files:
    Lx = int(f.split("_")[1])
    if Lx < 4:
        continue
    L_list.append(Lx)

    data_set = np.load(f)
    p_list = data_set['pg']

    for mode in ["mean", "var"]:
        key_mode = "mean" if mode=="mean" else "var"
        if key_mode not in TEE_data:
            TEE_data[key_mode] = {}

        for tee_idx in range(1,5):
            a = data_set[f'TEE{tee_idx}_ave'] if mode=="mean" else data_set[f'TEE{tee_idx}_var']
            da = data_set[f'TEE{tee_idx}_err'] * gamma

            # 端データ除外
            if d > 0:
                a = a[d:-d]
                da = da[d:-d]
                p_use = p_list[d:-d]
            else:
                p_use = p_list

            # データ格納
            key_tee = f'TEE{tee_idx}'
            if key_tee not in TEE_data[key_mode]:
                TEE_data[key_mode][key_tee] = {'a': [], 'da': [], 'p': p_use}
            TEE_data[key_mode][key_tee]['a'].append(a.tolist())
            TEE_data[key_mode][key_tee]['da'].append(da.tolist())

# --------------------------
# 元データプロット
# --------------------------
for mode in ["mean", "var"]:
    print(f"\n===== Original Data: {mode} =====")
    for tee_idx in range(1,5):
        data = TEE_data[mode][f'TEE{tee_idx}']
        p_use = data['p']
        fig, ax = plt.subplots(figsize=(5,4))
        for i, L in enumerate(L_list):
            ax.errorbar(p_use, data['a'][i], yerr=data['da'][i],
                        fmt=markers[i % len(markers)],
                        color=cmap((i+1)/8),
                        capsize=3, label=f"L={L}")
        ax.set_xlabel('p', fontsize=14)
        ax.set_ylabel('Average' if mode=="mean" else 'Variance', fontsize=14)
        ax.set_title(f'TEE{tee_idx} ({mode})')
        ax.legend()
        plt.show()

# --------------------------
# FSS 実行
# --------------------------
for mode in ["mean", "var"]:
    print(f"\n===== FSS: {mode} =====")
    for tee_idx in range(1,5):
        data = TEE_data[mode][f'TEE{tee_idx}']
        a = data['a']
        da = data['da']
        p_use = data['p']

        best_S = np.inf
        best_ret = None

        for ipc in pc_ini:
            try:
                ret = fssa.autoscale(L_list, p_use, a, da, ipc, 2, 2, x_bounds=None)
                xs, ys, dys = fssa.scaledata(L_list, p_use, a, da, ret.rho, ret.nu, ret.zeta)
                S = fssa.quality(xs, ys, dys, x_bounds=None)
                if S < best_S:
                    best_S = S
                    best_ret = ret
            except Exception as e:
                print(f"pc_ini={ipc} でエラー: {e}")

        if best_ret is None:
            print(f"TEE{tee_idx} は全ての初期値で失敗")
            continue

        # プロット（スケーリング後）
        auto_scaled_data = fssa.scaledata(L_list, p_use, a, da, best_ret.rho, best_ret.nu, best_ret.zeta)
        fig, ax = plt.subplots(figsize=(4,4))
        for i, L in enumerate(L_list):
            ax.scatter(auto_scaled_data.x.T[:,i], auto_scaled_data.y.T[:,i],
                       marker=markers[i % len(markers)],
                       color=cmap((i+1)/8), s=80)
        ax.set_xlabel(r'$(p-p_c)L^{1/\nu}$', fontsize=16)
        ax.set_ylabel(r'$F$', fontsize=16)
        ax.set_title(f'TEE{tee_idx} ({mode}) FSS')
        plt.show()

        # 結果出力
        print(f"TEE{tee_idx} pc: {best_ret.rho} ± {best_ret.drho}")
        print(f"TEE{tee_idx} nu: {best_ret.nu} ± {best_ret.dnu}")
        print(f"TEE{tee_idx} zeta: {best_ret.zeta} ± {best_ret.dzeta}")
        print(f"TEE{tee_idx} S (quality): {best_S}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import fssa
from cycler import cycler
import matplotlib.cm as cm

# --------------------------
# 設定
# --------------------------
L_list = [10, 20, 40]
p_list = np.linspace(0.6, 0.9, 21)
gamma = 0.2
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
cmap = cm.get_cmap('gnuplot')

TEE_data = {'a': [], 'da': [], 'p': p_list}

# --------------------------
# データ作成（ガウス型ピーク）
# --------------------------
pc_true = 0.75
nu_true = 1.0
zeta_true = 0.0

for L in L_list:
    # 真のスケーリング関数
    x = (p_list - pc_true) * L**(1/nu_true)
    F = np.exp(-x**2) + zeta_true
    # 分散と誤差を作る（大きい L は分散小）
    var = 0.05 / L
    err = np.sqrt(var)
    noise = np.random.normal(0, np.sqrt(var), size=p_list.shape)
    
    a = F + noise
    da = np.ones_like(a) * err * gamma
    
    TEE_data['a'].append(a)
    TEE_data['da'].append(da)

# --------------------------
# 元データプロット
# --------------------------
plt.figure(figsize=(5,4))
for i, L in enumerate(L_list):
    plt.errorbar(p_list, TEE_data['a'][i], yerr=TEE_data['da'][i], 
                 marker=markers[i], linestyle='', label=f'L={L}')
plt.xlabel('p')
plt.ylabel('F')
plt.title('Original Data')
plt.legend()
plt.show()

# --------------------------
# FSS 実行
# --------------------------
pc_ini = [0.74]  # 初期臨界点
a_list = TEE_data['a']
da_list = TEE_data['da']

ret = fssa.autoscale(L_list, p_list, a_list, da_list, pc_ini[0], 1.0, 1.0)
auto_scaled_data = fssa.scaledata(L_list, p_list, a_list, da_list, ret.rho, ret.nu, ret.zeta)

# --------------------------
# スケーリング後プロット
# --------------------------
plt.figure(figsize=(5,4))
for i, L in enumerate(L_list):
    plt.scatter(auto_scaled_data.x.T[:,i], auto_scaled_data.y.T[:,i],
                marker=markers[i], label=f'L={L}', s=80)
plt.xlabel(r'$(p-p_c)L^{1/\nu}$')
plt.ylabel('F')
plt.title('FSS Scaled Data')
plt.legend()
plt.show()

# --------------------------
# 結果出力
# --------------------------
print("FSS results:")
print("pc:", ret.rho, "±", ret.drho)
print("nu:", ret.nu, "±", ret.dnu)
print("zeta:", ret.zeta, "±", ret.dzeta)
```

```python
import numpy as np
import glob
import fssa
import matplotlib.pyplot as plt
from cycler import cycler
import matplotlib.cm as cm

# --------------------------
# 設定
# --------------------------
d = 0               # 端データ除外
gamma = 0.2         # 誤差スケーリング
markers = ['x', 'o', '^', 's', '*', 'H', 'D']
cmap = cm.get_cmap('gnuplot')
pc_ini = [0.80, 0.81, 0.82]  # 初期臨界点候補

# --------------------------
# 部分系サイズの例
# --------------------------
# L_list に対応する部分系サイズを設定
# 実際の部分系サイズに置き換えてください
N_sub = {
    1: 12, 2: 18, 3: 24, 4: 30
}

# --------------------------
# データ読み込み
# --------------------------
files = sorted(glob.glob("Lx_*_*.npz"))
L_list = [16,20,22,24]
TEE_data = {}  # {'mean':{}, 'var':{}}

for f in files:
    Lx = int(f.split("_")[1])
    if Lx < 4:
        continue
    L_list.append(Lx)

    data_set = np.load(f)
    p_list = data_set['pg']

    for tee_idx in range(1,5):
        a = data_set[f'TEE{tee_idx}_ave']  # 平均値を使用
        da = data_set[f'TEE{tee_idx}_err'] * gamma

        if d > 0:
            a = a[d:-d]
            da = da[d:-d]
            p_use = p_list[d:-d]
        else:
            p_use = p_list

        key_tee = f'TEE{tee_idx}'
        if key_tee not in TEE_data:
            TEE_data[key_tee] = {'a': [], 'da': [], 'p': p_use, 'L_sub': []}

        TEE_data[key_tee]['a'].append(a)
        TEE_data[key_tee]['da'].append(da)

        # 部分系サイズを追加（ここでは仮に Lx/2 などで例示）
        L_sub_val = Lx // 2
        TEE_data[key_tee]['L_sub'].append(L_sub_val)

# --------------------------
# 部分系サイズ依存 FSS 実行
# --------------------------
for tee_idx in range(1,5):
    data = TEE_data[f'TEE{tee_idx}']
    a = np.array(data['a'])          # shape = (N_L, Np)
    da = np.array(data['da'])
    L_eff = np.array(data['L_sub'])  # 部分系サイズ

    best_S = np.inf
    best_ret = None

    for pc0 in pc_ini:
        try:
            ret = fssa.autoscale(L_eff, data['p'], a, da, pc0, 1.0, 1.0)
            xs, ys, dys = fssa.scaledata(L_eff, data['p'], a, da, ret.rho, ret.nu, ret.zeta)
            S = fssa.quality(xs, ys, dys)
            if S < best_S:
                best_S = S
                best_ret = ret
        except Exception as e:
            print(f"pc_ini={pc0} でエラー: {e}")

    if best_ret is None:
        print(f"TEE{tee_idx} は全ての初期値で失敗")
        continue

    # スケーリング後プロット
    auto_scaled_data = fssa.scaledata(L_eff, data['p'], a, da, best_ret.rho, best_ret.nu, best_ret.zeta)
    fig, ax = plt.subplots(figsize=(4,4))
    for i, L in enumerate(L_eff):
        ax.scatter(auto_scaled_data.x.T[:,i], auto_scaled_data.y.T[:,i],
                   marker=markers[i % len(markers)],
                   color=cmap((i+1)/8), s=80)
    ax.set_xlabel(r'$(r-r_c)L_{\rm sub}^{1/\nu}$', fontsize=14)
    ax.set_ylabel(r'$F$', fontsize=14)
    plt.title(f'TEE{tee_idx} (部分系サイズ依存)')
    plt.show()

    # 結果出力
    print(f"TEE{tee_idx} pc: {best_ret.rho} ± {best_ret.drho}")
    print(f"TEE{tee_idx} nu: {best_ret.nu} ± {best_ret.dnu}")
    print(f"TEE{tee_idx} zeta: {best_ret.zeta} ± {best_ret.dzeta}")
    print(f"TEE{tee_idx} S (quality): {best_S}")
```
