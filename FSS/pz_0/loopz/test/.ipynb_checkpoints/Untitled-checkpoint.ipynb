```python
# ==== 最小ワンセル：slurm*.out 収集 → 長い表 → (L,pg)集計 ====
import re, glob, csv
import numpy as np
import pandas as pd

# 1) 収集対象（必要ならパターンを変えてください）
patterns = ["slurm-*.out"]  # 例: ["slurm-2660*.out", "slurm-1234.out"]

# 2) ゆるめの正規表現（小数/指数/空白の揺れ対応）
PAT = re.compile(
    r"Lx\s*=\s*Ly\s*=\s*(?P<L>\d+).*?"
    r"pg\s*=\s*(?P<pg>[0-9.+\-Ee]+).*?"
    r"R2z[_ ]?loop\s*=\s*(?P<r2>[0-9.+\-Ee]+)"
)

rows = []
files = []
for pat in patterns:
    files.extend(glob.glob(pat))

for fn in sorted(files):
    with open(fn, "r", encoding="utf-8", errors="ignore") as f:
        for ln in f:
            m = PAT.search(ln)
            if not m:
                continue
            rows.append((int(m["L"]), float(m["pg"]), float(m["r2"])))

# 3) 長い表をCSV出力（L,pg,R2z_loop）
if rows:
    with open("r2z_long.csv", "w", newline="", encoding="utf-8") as fw:
        wr = csv.writer(fw); wr.writerow(["L","pg","R2z_loop"]); wr.writerows(rows)

# 4) (L,pg)ごとに 平均・分散(標本)・n を計算 → CSV出力
if rows:
    df = pd.DataFrame(rows, columns=["L","pg","R2z_loop"])
    summary = df.groupby(["L","pg"])["R2z_loop"].agg(
        mean="mean",
        var=lambda x: np.var(x, ddof=1),
        n="count"
    ).reset_index()
    summary.sort_values(["L","pg"]).to_csv("r2z_var_summary.csv", index=False)

# 5) ざっくり状況表示
print(f"読み込んだファイル数: {len(files)}")
print(f"抽出した行数: {len(rows)}")
if rows:
    print("L 種類:", sorted(df["L"].unique().tolist()))
    print("pg 範囲:", float(df["pg"].min()), "〜", float(df["pg"].max()))
    print("\n(L,pg)ごとのサンプル数 先頭10行:")
    print(df.groupby(["L","pg"]).size().reset_index(name="n").sort_values(["L","pg"]).head(10).to_string(index=False))
    print("\nsummary 先頭5行:")
    print(summary.head().to_string(index=False))
else:
    print("※ 行が抽出できませんでした。パターンや正規表現を見直してください。")
```

```python
import pandas as pd
import numpy as np

# 1) (L,pg)ごとの集計結果を読み込み
summary = pd.read_csv("r2z_var_summary.csv")  # 列: L, pg, mean, var, n

# 2) 分散の標準誤差（ガウス近似）を追加: SE[S^2] ≈ S^2 * sqrt(2/(n-1))
summary["se_var"] = summary.apply(
    lambda r: r["var"] * np.sqrt(2.0 / (r["n"] - 1)) if r["n"] > 1 else np.nan,
    axis=1
)

# （任意）過小誤差対策の簡易フロア：必要なければコメントアウト
EPS_REL = 0.1   # 相対フロア（7%）
EPS_ABS = 1e-3   # 絶対フロア
summary["se_eff"] = np.maximum(summary["se_var"], np.maximum(EPS_REL*np.abs(summary["var"]), EPS_ABS))

# 3) pyfssa 用の辞書 {L: (pg, var, se)} を作成
data_by_L = {}
for L, sub in summary.sort_values(["L","pg"]).groupby("L"):
    pg  = sub["pg"].to_numpy()
    var = sub["var"].to_numpy()
    se  = sub["se_eff"].to_numpy()   # フロアを使う。素の se_var を使いたければここを置き換え
    data_by_L[int(L)] = (pg, var, se)

# 4) 保存（npzに {str(L): (pg, var, se)} で格納）
np.savez("r2z_var_for_pyfssa.npz", **{str(k): v for k, v in data_by_L.items()})
print("[OK] r2z_var_for_pyfssa.npz を保存しました。L =", sorted(data_by_L.keys()))

# 5) さっと中身を確認（各Lで先頭5点）
z = np.load("r2z_var_for_pyfssa.npz", allow_pickle=True)
for key in sorted(z.files, key=lambda s: int(s)):
    pg, var, se = z[key]
    print(f"\n=== L = {key} ===")
    for i in range(min(5, len(pg))):
        print(f"pg={pg[i]:.3f}, var={var[i]:.4e}, se={se[i]:.4e}")
    print(f"... total {len(pg)} pts")
```

```python
# ==== R2z_loop の分散FSS（autoscale/scaledata/quality最小版） ====
import numpy as np
import matplotlib.pyplot as plt
import fssa

# 0) 事前に作った {str(L): (pg, var, se)} をロード
ZFN = "r2z_var_for_pyfssa.npz"   # ←あなたの前段で保存したファイル
z = np.load(ZFN, allow_pickle=True)

# ---- L順に並べ、共通 pg グリッドを作る（全Lの共通集合を使用） ----
all_Ls = sorted(int(k) for k in z.files)

Ls_keep = [12,14, 16, 18, 20,22]   # ←使いたい L を列挙（例）
Ls_sorted = [L for L in all_Ls if L in Ls_keep]
pg_lists = [z[str(L)][0] for L in Ls_sorted]

# ほぼ同じ点同士を揃えるため小さな丸めで共通集合を取る
def to_key(arr, nd=6): return tuple(np.round(arr.astype(float), nd))
pg_sets = [set(to_key(p)) for p in pg_lists]
pg_common = sorted(list(set.intersection(*pg_sets)))
if len(pg_common) < 5:
    raise ValueError("共通の pg 点が少なすぎます。ログのpgを揃えるか、補間が必要です。")

pg_common = np.array(pg_common, dtype=float)
nL, nP = len(Ls_sorted), len(pg_common)

# 行列化: a_var[L_i, j], da_var[L_i, j] を共通pg順に揃える
a_var   = np.empty((nL, nP), float)
da_var  = np.empty((nL, nP), float)
for i, L in enumerate(Ls_sorted):
    pg, var, se = z[str(L)]
    # pg -> index マップ
    key2idx = {k: idx for idx, k in enumerate(to_key(pg))}
    idxs = [key2idx[k] for k in to_key(pg_common)]
    a_var[i, :]  = var[idxs]
    da_var[i, :] = se[idxs]

# （任意）pの使用範囲を絞りたければここでマスク

# フル配列（共通グリッド）から一発で絞る
p_full  = pg_common          # shape = (N_full,)
Y_full  = a_var              # shape = (nL, N_full)
DY_full = da_var             # shape = (nL, N_full)

# 端を落とす範囲を指定（例：0.70〜0.84まで残す＝それ以外は除外）
pmin, pmax = 0.7, 0.9
m = (p_full >= pmin) & (p_full <= pmax)

# ここで“列”をまとめて間引く（全Lに同じマスク）
p_use = p_full[m]
y     = Y_full[:,  m]
dy    = DY_full[:, m]

# 整合チェック（全部 True ならOK）
assert y.shape[1] == dy.shape[1] == p_use.size
print(f"[info] p range = [{p_use.min():.3f}, {p_use.max():.3f}] (N={p_use.size})")

print(f"[info] Ls={Ls_sorted}")
print(f"[info] p range = [{p_use.min():.3f}, {p_use.max():.3f}] (N={len(p_use)})")

# 1) autoscale（感受率型：zeta が γ/ν に相当）
#    初期値はピーク中央値→pc、ν~1.0、γ/ν~1.0 あたりから
def guess_pc_from_peaks(y_matrix, p_vec):
    # 各Lで分散最大のpgを取り、その中央値
    pcs = []
    for i in range(y_matrix.shape[0]):
        j = int(np.argmax(y_matrix[i]))
        pcs.append(p_vec[j])
    return float(np.median(pcs))

#pc0   = guess_pc_from_peaks(y, p_use)
pc0   = 0.785
nu0   = 1.53
zeta0 = -0.2   # γ/ν の初期値

ret = fssa.autoscale(Ls_sorted, p_use, y, dy, rho_c0=pc0, nu0=nu0, zeta0=zeta0)

# 2) 縮約後の座標を作る
xs, ys, dys = fssa.scaledata(Ls_sorted, p_use, y, dy, ret.rho, ret.nu, ret.zeta)

S = fssa.quality(xs, ys, dys)

# 3) プロット（前後）と quality
# 前：そのまま Var vs p
plt.figure(figsize=(6,4.5))
for i, L in enumerate(Ls_sorted):
    plt.plot(p_use, y[i], 'o-', label=f"L={L}")
plt.xlabel("pg"); plt.ylabel("Var[R2z_loop]")
plt.title("Before collapse"); plt.grid(True); plt.legend()
plt.show()

# 後：縮約後（xs,ys は shape = (nP, nL) を想定 → 転置で列=各L）
plt.figure(figsize=(6,4.5))
for i, L in enumerate(Ls_sorted):
    plt.plot(xs.T[:, i], ys.T[:, i], 'o', label=f"L={L}")
plt.xlabel(r"$(p-p_c)L^{1/\nu}$"); plt.ylabel(r"$\mathrm{Var}\,L^{-\gamma/\nu}$")
plt.title("After collapse"); plt.grid(True); plt.legend()
plt.show()


# 4) 結果表示
print("=== Variance FSS (z-loop) ===")
print(f"pc       = {ret.rho:.6f}  ± {ret.drho:.6f}")
print(f"nu       = {ret.nu:.6f}   ± {ret.dnu:.6f}")
print(f"zeta = {ret.zeta:.6f} ± {ret.dzeta:.6f}")
print(f"quality S = {S:.6g}")
```
