```python
# -*- coding: utf-8 -*-
"""
Final FSS (variance only) for TWO datasets, with optional manual p-window and pooled pc:
1) p-window: auto by variance peak ±dp (fallback K-nearest) → optional MANUAL_WINDOW override
2) zeta profiling (coarse→refine) → (pc*, nu*, zeta*, S*)
3) FAST p-LOO jackknife with zeta* fixed → pc, nu ± SE
4) Side-by-side summary and pooled pc (inverse-variance weighted)

Edit DATASETS, TEE_KEYS, L_LIST, and optionally MANUAL_WINDOW.
"""

import numpy as np
import warnings
import fssa

# ============ CONFIG ============
# 解析対象（2つ以上でもOK）
DATASETS = [
    ("Lx_24_Nd_500_NT_17280_px_0.0_MPI_hexagonal6.npz", "A"),
    ("Lx_24_Nd_500_NT_17280_pz_0.0_MPI_hexagonal6.npz", "B"),
]

# L キー（npz 内の系列名）と L の値
TEE_KEYS = ['TEE2','TEE3','TEE4','TEE5','TEE6']
L_LIST   = [np.sqrt(36), np.sqrt(54), np.sqrt(72), np.sqrt(90), np.sqrt(108)]

# --- 手動窓（使うなら (0.81, 0.89) などを設定、オフなら None）---
MANUAL_WINDOW = (0.75, 0.9)   # ←固定したいならここを設定。自動窓にしたいときは None

# 自動窓の設定（分散ピーク±dp → 足りなければ K-nearest）
DP_TRY    = [0.04, 0.05, 0.06]
MIN_PTS   = 9
K_NEAREST = 13

# 探索レンジ
NU_RANGE   = (1.0, 3.0)
ZETA_RANGE = (0.5, 4.0)  # ζ >= 0.5 で安定

# グリッド解像度（速度と精度のバランス）
RHO_STEPS_COARSE  = 31
NU_STEPS_COARSE   = 31
ZETA_STEPS_COARSE = 25
RHO_HALF_REF  = 0.010; RHO_STEPS_REF = 31
NU_HALF_REF   = 0.300; NU_STEPS_REF  = 31
ZETA_HALF_REF = 0.400; ZETA_STEPS_REF = 31

# FAST ジャックナイフ（ζ固定で (pc,nu) 最小化）
RHO_STEPS_C_FAST = 31
NU_STEPS_C_FAST  = 31
RHO_HALF_REF_FAST = 0.010; RHO_STEPS_R_FAST = 31
NU_HALF_REF_FAST  = 0.300; NU_STEPS_R_FAST  = 31

# 誤差の扱い
GAMMA_ERR = 0.5   # *_err に掛ける係数
ERR_FLOOR = 1e-6  # 絶対床（ゼロ/極小重みを防ぐ）

warnings.filterwarnings("ignore", category=RuntimeWarning)

# ============ UTILS ============
def ensure_sorted_unique_p(p, arrays_on_p):
    p = np.asarray(p).ravel()
    idx = np.argsort(p)
    p_sorted = p[idx]
    arrays_sorted = [A[:, idx] for A in arrays_on_p]
    uniq, inv, cnt = np.unique(p_sorted, return_inverse=True, return_counts=True)
    if uniq.size != p_sorted.size:
        new_arrays = []
        for A in arrays_sorted:
            out = np.zeros((A.shape[0], uniq.size), dtype=A.dtype)
            for g in range(uniq.size):
                cols = (inv == g)
                out[:, g] = A[:, cols].mean(axis=1)
            new_arrays.append(out)
        return uniq, new_arrays
    return p_sorted, arrays_sorted

def pick_window_by_peak(p_all, a_var, dp_list, min_pts, k_nearest):
    avg = a_var.mean(axis=0)
    i_peak = int(np.argmax(avg))
    p_peak = float(p_all[i_peak])
    for dp in dp_list:
        mask = (p_all >= p_peak - dp) & (p_all <= p_peak + dp)
        if mask.sum() >= min_pts:
            return mask, p_peak, ("±dp", dp)
    order = np.argsort(np.abs(p_all - p_peak))
    sel   = np.sort(order[:min(k_nearest, p_all.size)])
    mask  = np.zeros_like(p_all, dtype=bool); mask[sel] = True
    return mask, p_peak, ("K-nearest", int(mask.sum()))

def eval_quality(Ls, p, A, dA, rho, nu, zeta):
    try:
        xs, ys, dys = fssa.scaledata(Ls, p, A, dA, rho, nu, zeta)
        if not (np.isfinite(xs).all() and np.isfinite(ys).all() and np.isfinite(dys).all()):
            return np.inf
        S = fssa.quality(xs, ys, dys)
        return (np.inf if not np.isfinite(S) else S)
    except Exception:
        return np.inf

def grid_min_fixed_zeta(Ls, p, A, dA, zeta, rho_grid, nu_grid):
    best = (np.inf, None, None)
    for rho0 in rho_grid:
        for nu0 in nu_grid:
            S = eval_quality(Ls, p, A, dA, rho0, nu0, zeta)
            if S < best[0]:
                best = (S, rho0, nu0)
    return best  # S, rho, nu

def linspace_clip(center, half, steps, lo, hi):
    a = max(lo, center - half); b = min(hi, center + half)
    if a == b:
        eps = max(half, 1e-3); a = max(lo, center - eps); b = min(hi, center + eps)
    return np.linspace(a, b, steps)

def profile_once(Ls, p, A, dA, verbose=False):
    # coarse
    rho_grid_c = np.linspace(p.min(), p.max(), RHO_STEPS_COARSE)
    nu_grid_c  = np.linspace(NU_RANGE[0], NU_RANGE[1], NU_STEPS_COARSE)
    z_grid_c   = np.linspace(ZETA_RANGE[0], ZETA_RANGE[1], ZETA_STEPS_COARSE)
    best = (np.inf, None, None, None)
    for j, z0 in enumerate(z_grid_c, 1):
        if verbose and (j in (1, len(z_grid_c)) or j % max(1, len(z_grid_c)//5) == 0):
            print(f"  [coarse] zeta {j}/{len(z_grid_c)} z0={z0:.3f}")
        Smin, rho_b, nu_b = grid_min_fixed_zeta(Ls, p, A, dA, z0, rho_grid_c, nu_grid_c)
        if Smin < best[0]:
            best = (Smin, rho_b, nu_b, z0)
    S_c, rho_c, nu_c, zeta_c = best
    # refine
    rho_grid_r = linspace_clip(rho_c,  RHO_HALF_REF,  RHO_STEPS_REF, p.min(), p.max())
    nu_grid_r  = linspace_clip(nu_c,   NU_HALF_REF,   NU_STEPS_REF,  *NU_RANGE)
    z_grid_r   = linspace_clip(zeta_c, ZETA_HALF_REF, ZETA_STEPS_REF, *ZETA_RANGE)
    best_r = (np.inf, None, None, None)
    for z0 in z_grid_r:
        Smin, rho_b, nu_b = grid_min_fixed_zeta(Ls, p, A, dA, z0, rho_grid_r, nu_grid_r)
        if Smin < best_r[0]:
            best_r = (Smin, rho_b, nu_b, z0)
    S, rho, nu, zeta = best_r
    return dict(pc=rho, nu=nu, zeta=zeta, S=S)

def jackknife_se(vals):
    vals = np.asarray(vals, float)
    n = vals.size
    mean = vals.mean()
    se = np.sqrt((n-1)/n * np.sum((vals - mean)**2))
    return mean, se

def min_pc_nu_given_zeta(Ls, p, A, dA, zeta):
    # coarse
    rho_grid_c = np.linspace(p.min(), p.max(), RHO_STEPS_C_FAST)
    nu_grid_c  = np.linspace(NU_RANGE[0], NU_RANGE[1], NU_STEPS_C_FAST)
    best_c = (np.inf, None, None)
    for rho0 in rho_grid_c:
        for nu0 in nu_grid_c:
            S = eval_quality(Ls, p, A, dA, rho0, nu0, zeta)
            if S < best_c[0]:
                best_c = (S, rho0, nu0)
    _, rho_c, nu_c = best_c
    # refine
    rho_grid_r = linspace_clip(rho_c, RHO_HALF_REF_FAST, RHO_STEPS_R_FAST, p.min(), p.max())
    nu_grid_r  = linspace_clip(nu_c,  NU_HALF_REF_FAST,  NU_STEPS_R_FAST,  *NU_RANGE)
    best_r = (np.inf, None, None)
    for rho0 in rho_grid_r:
        for nu0 in nu_grid_r:
            S = eval_quality(Ls, p, A, dA, rho0, nu0, zeta)
            if S < best_r[0]:
                best_r = (S, rho0, nu0)
    S_star, rho_star, nu_star = best_r
    return dict(pc=rho_star, nu=nu_star, S=S_star)

def zeta_se_by_parabola(Ls, p, A, dA, z_star, dz=0.2, steps=9):
    zs = np.linspace(z_star - dz, z_star + dz, steps)
    Ss = []
    rho_grid_c = np.linspace(p.min(), p.max(), 21)
    nu_grid_c  = np.linspace(NU_RANGE[0], NU_RANGE[1], 21)
    for z0 in zs:
        Smin = np.inf
        for rho0 in rho_grid_c:
            for nu0 in nu_grid_c:
                S = eval_quality(Ls, p, A, dA, rho0, nu0, z0)
                if S < Smin:
                    Smin = S
        Ss.append(Smin)
    zs = np.asarray(zs); Ss = np.asarray(Ss)
    coef = np.polyfit(zs - z_star, Ss, 2)  # S ≈ a (z - z*)^2 + c
    a = coef[0]
    if not np.isfinite(a) or a <= 0:
        return 0.0
    dS = 0.02  # 目安（必要に応じ 0.01–0.05）
    return np.sqrt(dS / a)

# ============ RUNNER ============
def run_one_dataset(npz_path, label):
    print(f"\n==================== {label} ====================")
    data  = np.load(npz_path)
    p_all = data['pg'].astype(float)
    a_var = np.array([data[f'{k}_var'] for k in TEE_KEYS], dtype=float)
    dA0   = np.array([data[f'{k}_err'] for k in TEE_KEYS], dtype=float) * GAMMA_ERR

    # p 整形（昇順・重複pを平均で潰す）
    p_all, [a_var, dA0] = ensure_sorted_unique_p(p_all, [a_var, dA0])
    # 誤差フロア
    dA0 = np.maximum(dA0, ERR_FLOOR * np.maximum(1.0, np.abs(a_var)))

    # Step1: 自動窓
    mask, p_peak, how = pick_window_by_peak(p_all, a_var, DP_TRY, MIN_PTS, K_NEAREST)
    p  = p_all[mask]; A = a_var[:, mask]; dA = dA0[:, mask]
    print(f"[window:auto] mode={how[0]}, arg={how[1]} | peak≈{p_peak:.6f}, "
          f"range=[{p.min():.6f},{p.max():.6f}], Np={p.size}")

    # --- 手動窓が設定されていれば上書き ---
    if MANUAL_WINDOW is not None:
        pmin, pmax = MANUAL_WINDOW
        mask_m = (p_all >= pmin) & (p_all <= pmax)
        if mask_m.sum() >= MIN_PTS:
            p, A, dA = p_all[mask_m], a_var[:, mask_m], dA0[:, mask_m]
            print(f"[window:manual] range=[{p.min():.6f},{p.max():.6f}], Np={p.size}")
        else:
            print(f"[window:manual] insufficient points ({mask_m.sum()}) → using auto window.")

    # Step2: ζプロファイル
    print("[baseline] profiling zeta...")
    base = profile_once(L_LIST, p, A, dA, verbose=True)
    print("  -> pc={:.6f}, nu={:.6f}, zeta={:.6f}, S={:.6f}".format(
        base['pc'], base['nu'], base['zeta'], base['S']))

    # Step3: FAST p-LOO（ζ固定）
    z_fix = float(base['zeta'])
    pcs, nus = [], []
    N = p.size
    print("[jackknife-fast] zeta fixed = {:.6f} | LOO p x{}".format(z_fix, N))
    for j in range(N):
        mask_j = np.ones(N, dtype=bool); mask_j[j] = False
        pj, Aj, dAj = p[mask_j], A[:, mask_j], dA[:, mask_j]
        est = min_pc_nu_given_zeta(L_LIST, pj, Aj, dAj, z_fix)
        pcs.append(est['pc']); nus.append(est['nu'])

    # 集計
    pc_mean, pc_se = jackknife_se(pcs)
    nu_mean, nu_se = jackknife_se(nus)
    zeta_se = zeta_se_by_parabola(L_LIST, p, A, dA, z_fix, dz=0.2, steps=9)

    print("\n=== RESULT [{}] ===".format(label))
    print("pc   = {:.6f} ± {:.6f}".format(pc_mean, pc_se))
    print("nu   = {:.6f} ± {:.6f}".format(nu_mean, nu_se))
    print("zeta = {:.6f} ± {:.6f}  (parabola approx)".format(z_fix, zeta_se))
    print("S*   = {:.6f}  (baseline min S)".format(base['S']))

    return dict(label=label, pc=pc_mean, dpc=pc_se,
                nu=nu_mean, dnu=nu_se, zeta=z_fix, dzeta=zeta_se, S=base['S'])

# ============ MAIN ============
all_results = []
for path, label in DATASETS:
    all_results.append(run_one_dataset(path, label))

# 並べて表示
print("\n==================== SUMMARY (side-by-side) ====================")
print("{:>6s} | {:>10s} ± {:>9s} | {:>10s} ± {:>9s} | {:>10s} ± {:>9s} | {:>6s}".format(
    "label", "pc", "dpc", "nu", "dnu", "zeta", "dzeta", "S*"))
for r in all_results:
    print("{:>6s} | {:10.6f} ± {:9.6f} | {:10.6f} ± {:9.6f} | {:10.6f} ± {:9.6f} | {:6.4f}".format(
        r['label'], r['pc'], r['dpc'], r['nu'], r['dnu'], r['zeta'], r['dzeta'], r['S']))
print("================================================================")

# プール推定（逆分散重み）
pcs = np.array([r['pc']  for r in all_results], float)
ses = np.array([r['dpc'] for r in all_results], float)
valid = np.isfinite(pcs) & np.isfinite(ses) & (ses > 0)
if valid.sum() >= 1:
    w = 1.0 / (ses[valid]**2)
    pc_pooled = np.sum(w * pcs[valid]) / np.sum(w)
    pc_pooled_se = np.sqrt(1.0 / np.sum(w))
    print("\n==================== POOLED ESTIMATE ====================")
    print(f"pc (pooled) = {pc_pooled:.6f}  ±  {pc_pooled_se:.6f}")
    print("=========================================================")
else:
    print("\n[pooled] insufficient valid entries to compute pooled pc.")
```

```python
# -*- coding: utf-8 -*-
"""
Two-dataset FSS (variance only) with per-dataset windows and L trimming:
- A: use manual p-window ~[0.77,0.85], drop smallest L to reduce non-asymptotic bias
- B: use manual p-window [0.75,0.90] as you used before
- Zeta profiling (coarse→refine), then FAST p-LOO jackknife with zeta* fixed
- Warn if zeta hits search bounds
- Side-by-side summary; do NOT pool unless both S<=0.1 and pc overlap
"""

import numpy as np
import warnings
import fssa

# ===== CONFIG =====
DATASETS = [
    ("Lx_24_Nd_500_NT_17280_px_0.0_MPI_hexagonal6.npz", "A"),
    ("Lx_24_Nd_500_NT_17280_pz_0.0_MPI_hexagonal6.npz", "B"),
]

TEE_KEYS = ['TEE2','TEE3','TEE4','TEE5','TEE6']
L_LIST   = [np.sqrt(36), np.sqrt(54), np.sqrt(72), np.sqrt(90), np.sqrt(108)]

# データ別の手動窓（None で自動窓を使用）
MANUAL_WINDOWS = {
    "A": (0.77, 0.90),   # Aは狭く（autoが示した範囲）
    "B": (0.75, 0.90),   # Bは従来の広めの窓でOK
}

# データ別に「最小Lを外す」指定
DROP_SMALLEST_L = {
    "A": True,   # Aは最小サイズを外して安定化
    "B": False,
}

# 自動窓の規則（使う場面があれば）
DP_TRY, MIN_PTS, K_NEAREST = [0.04, 0.05, 0.06], 9, 13

# 探索レンジ（Aが上限4.0に張り付いたので、少し現実的に絞るのも手）
NU_RANGE   = (1.0, 3.0)
ZETA_RANGE = (0.5, 3.5)   # 3.5 に下げ、上限張り付き抑止（必要なら戻せる）

# グリッド解像度
RHO_STEPS_COARSE, NU_STEPS_COARSE, ZETA_STEPS_COARSE = 31, 31, 25
RHO_HALF_REF, RHO_STEPS_REF = 0.010, 31
NU_HALF_REF,  NU_STEPS_REF  = 0.300, 31
ZETA_HALF_REF, ZETA_STEPS_REF = 0.400, 31

# FAST jackknife (ζ固定で (pc,ν))
RHO_STEPS_C_FAST, NU_STEPS_C_FAST = 31, 31
RHO_HALF_REF_FAST, RHO_STEPS_R_FAST = 0.010, 31
NU_HALF_REF_FAST,  NU_STEPS_R_FAST  = 0.300, 31

# 誤差
GAMMA_ERR, ERR_FLOOR = 0.5, 1e-6

warnings.filterwarnings("ignore", category=RuntimeWarning)

# ===== UTILS =====
def ensure_sorted_unique_p(p, arrays_on_p):
    p = np.asarray(p).ravel()
    idx = np.argsort(p)
    p_sorted = p[idx]
    arrays_sorted = [A[:, idx] for A in arrays_on_p]
    uniq, inv, cnt = np.unique(p_sorted, return_inverse=True, return_counts=True)
    if uniq.size != p_sorted.size:
        new_arrays = []
        for A in arrays_sorted:
            out = np.zeros((A.shape[0], uniq.size), dtype=A.dtype)
            for g in range(uniq.size):
                cols = (inv == g)
                out[:, g] = A[:, cols].mean(axis=1)
            new_arrays.append(out)
        return uniq, new_arrays
    return p_sorted, arrays_sorted

def pick_window_by_peak(p_all, a_var):
    avg = a_var.mean(axis=0)
    i_peak = int(np.argmax(avg)); p_peak = float(p_all[i_peak])
    for dp in DP_TRY:
        mask = (p_all >= p_peak - dp) & (p_all <= p_peak + dp)
        if mask.sum() >= MIN_PTS:
            return mask, ("±dp", dp), p_peak
    order = np.argsort(np.abs(p_all - p_peak))
    sel   = np.sort(order[:min(K_NEAREST, p_all.size)])
    mask  = np.zeros_like(p_all, dtype=bool); mask[sel] = True
    return mask, ("K-nearest", int(mask.sum())), p_peak

def eval_quality(Ls, p, A, dA, rho, nu, zeta):
    try:
        xs, ys, dys = fssa.scaledata(Ls, p, A, dA, rho, nu, zeta)
        if not (np.isfinite(xs).all() and np.isfinite(ys).all() and np.isfinite(dys).all()):
            return np.inf
        S = fssa.quality(xs, ys, dys)
        return (np.inf if not np.isfinite(S) else S)
    except Exception:
        return np.inf

def grid_min_fixed_zeta(Ls, p, A, dA, zeta, rho_grid, nu_grid):
    best = (np.inf, None, None)
    for rho0 in rho_grid:
        for nu0 in nu_grid:
            S = eval_quality(Ls, p, A, dA, rho0, nu0, zeta)
            if S < best[0]:
                best = (S, rho0, nu0)
    return best

def linspace_clip(center, half, steps, lo, hi):
    a = max(lo, center - half); b = min(hi, center + half)
    if a == b:
        eps = max(half, 1e-3); a = max(lo, center - eps); b = min(hi, center + eps)
    return np.linspace(a, b, steps)

def profile_once(Ls, p, A, dA, verbose=False):
    rho_grid_c = np.linspace(p.min(), p.max(), RHO_STEPS_COARSE)
    nu_grid_c  = np.linspace(NU_RANGE[0], NU_RANGE[1], NU_STEPS_COARSE)
    z_grid_c   = np.linspace(ZETA_RANGE[0], ZETA_RANGE[1], ZETA_STEPS_COARSE)
    best = (np.inf, None, None, None)
    for j, z0 in enumerate(z_grid_c, 1):
        if verbose and (j in (1, len(z_grid_c)) or j % max(1, len(z_grid_c)//5) == 0):
            print(f"  [coarse] zeta {j}/{len(z_grid_c)} z0={z0:.3f}")
        Smin, rho_b, nu_b = grid_min_fixed_zeta(Ls, p, A, dA, z0, rho_grid_c, nu_grid_c)
        if Smin < best[0]:
            best = (Smin, rho_b, nu_b, z0)
    S_c, rho_c, nu_c, zeta_c = best
    rho_grid_r = linspace_clip(rho_c,  RHO_HALF_REF,  RHO_STEPS_REF, p.min(), p.max())
    nu_grid_r  = linspace_clip(nu_c,   NU_HALF_REF,   NU_STEPS_REF,  *NU_RANGE)
    z_grid_r   = linspace_clip(zeta_c, ZETA_HALF_REF, ZETA_STEPS_REF, *ZETA_RANGE)
    best_r = (np.inf, None, None, None)
    for z0 in z_grid_r:
        Smin, rho_b, nu_b = grid_min_fixed_zeta(Ls, p, A, dA, z0, rho_grid_r, nu_grid_r)
        if Smin < best_r[0]:
            best_r = (Smin, rho_b, nu_b, z0)
    S, rho, nu, zeta = best_r
    return dict(pc=rho, nu=nu, zeta=zeta, S=S)

def jackknife_se(vals):
    vals = np.asarray(vals, float)
    n = vals.size
    mean = vals.mean()
    se = np.sqrt((n-1)/n * np.sum((vals - mean)**2))
    return mean, se

def min_pc_nu_given_zeta(Ls, p, A, dA, zeta):
    rho_grid_c = np.linspace(p.min(), p.max(), RHO_STEPS_C_FAST)
    nu_grid_c  = np.linspace(NU_RANGE[0], NU_RANGE[1], NU_STEPS_C_FAST)
    best_c = (np.inf, None, None)
    for rho0 in rho_grid_c:
        for nu0 in nu_grid_c:
            S = eval_quality(Ls, p, A, dA, rho0, nu0, zeta)
            if S < best_c[0]:
                best_c = (S, rho0, nu0)
    _, rho_c, nu_c = best_c
    rho_grid_r = linspace_clip(rho_c, RHO_HALF_REF_FAST, RHO_STEPS_R_FAST, p.min(), p.max())
    nu_grid_r  = linspace_clip(nu_c,  NU_HALF_REF_FAST,  NU_STEPS_R_FAST,  *NU_RANGE)
    best_r = (np.inf, None, None)
    for rho0 in rho_grid_r:
        for nu0 in nu_grid_r:
            S = eval_quality(Ls, p, A, dA, rho0, nu0, zeta)
            if S < best_r[0]:
                best_r = (S, rho0, nu0)
    S_star, rho_star, nu_star = best_r
    return dict(pc=rho_star, nu=nu_star, S=S_star)

def zeta_se_by_parabola(Ls, p, A, dA, z_star, dz=0.3, steps=13):
    zs = np.linspace(z_star - dz, z_star + dz, steps)
    Ss = []
    rho_grid_c = np.linspace(p.min(), p.max(), 21)
    nu_grid_c  = np.linspace(NU_RANGE[0], NU_RANGE[1], 21)
    for z0 in zs:
        Smin = np.inf
        for rho0 in rho_grid_c:
            for nu0 in nu_grid_c:
                S = eval_quality(Ls, p, A, dA, rho0, nu0, z0)
                if S < Smin: Smin = S
        Ss.append(Smin)
    zs = np.asarray(zs); Ss = np.asarray(Ss)
    coef = np.polyfit(zs - z_star, Ss, 2)  # S ≈ a (z - z*)^2 + c
    a = coef[0]
    if not np.isfinite(a) or a <= 0:  # 曲率がない/負なら推定不能
        return 0.0
    dS = 0.02
    return np.sqrt(dS / a)

# ===== RUNNER =====
def run_one_dataset(npz_path, label):
    print(f"\n==================== {label} ====================")
    d     = np.load(npz_path)
    p_all = d['pg'].astype(float)
    a_var = np.array([d[f'{k}_var'] for k in TEE_KEYS], float)
    dA0   = np.array([d[f'{k}_err'] for k in TEE_KEYS], float) * GAMMA_ERR

    p_all, [a_var, dA0] = ensure_sorted_unique_p(p_all, [a_var, dA0])
    dA0 = np.maximum(dA0, ERR_FLOOR * np.maximum(1.0, np.abs(a_var)))

    # 窓（自動）
    mask_auto, how, p_peak = pick_window_by_peak(p_all, a_var)
    p, A, dA = p_all[mask_auto], a_var[:, mask_auto], dA0[:, mask_auto]
    print(f"[window:auto] mode={how[0]}, arg={how[1]} | peak≈{p_peak:.6f}, "
          f"range=[{p.min():.6f},{p.max():.6f}], Np={p.size}")

    # 窓（手動、あれば上書き）
    if label in MANUAL_WINDOWS and MANUAL_WINDOWS[label] is not None:
        pmin, pmax = MANUAL_WINDOWS[label]
        m = (p_all >= pmin) & (p_all <= pmax)
        if m.sum() >= MIN_PTS:
            p, A, dA = p_all[m], a_var[:, m], dA0[:, m]
            print(f"[window:manual] range=[{p.min():.6f},{p.max():.6f}], Np={p.size}")
        else:
            print(f"[window:manual] insufficient points ({m.sum()}) → keep auto window.")

    # L のトリミング
    Ls_use = list(L_LIST)
    if DROP_SMALLEST_L.get(label, False):
        Ls_use = Ls_use[1:]
        A, dA = A[1:, :], dA[1:, :]
        print("[sizes] dropped smallest L for stability")

    # ζプロファイル
    print("[baseline] profiling zeta...")
    base = profile_once(Ls_use, p, A, dA, verbose=True)
    print("  -> pc={:.6f}, nu={:.6f}, zeta={:.6f}, S={:.6f}".format(
        base['pc'], base['nu'], base['zeta'], base['S']))

    # 境界チェック（zeta が範囲端なら警告）
    eps = 1e-8
    if abs(base['zeta'] - ZETA_RANGE[0]) < eps or abs(base['zeta'] - ZETA_RANGE[1]) < eps:
        print("[warn] zeta hit the search boundary. Consider narrowing window or revising ZETA_RANGE.")

    # FAST p-LOO（ζ固定）
    z_fix = float(base['zeta'])
    pcs, nus = [], []
    Np = p.size
    print(f"[jackknife-fast] zeta fixed = {z_fix:.6f} | LOO p x{Np}")
    for j in range(Np):
        mask_j = np.ones(Np, dtype=bool); mask_j[j] = False
        pj, Aj, dAj = p[mask_j], A[:, mask_j], dA[:, mask_j]
        est = min_pc_nu_given_zeta(Ls_use, pj, Aj, dAj, z_fix)
        pcs.append(est['pc']); nus.append(est['nu'])

    # 集計
    def jk(vals):
        v = np.asarray(vals, float); n = v.size
        m = v.mean(); se = np.sqrt((n-1)/n * np.sum((v-m)**2))
        return m, se

    pc_mean, pc_se = jk(pcs)
    nu_mean, nu_se = jk(nus)
    zeta_se = zeta_se_by_parabola(Ls_use, p, A, dA, z_fix, dz=0.3, steps=13)

    print("\n=== RESULT [{}] ===".format(label))
    print("pc   = {:.6f} ± {:.6f}".format(pc_mean, pc_se))
    print("nu   = {:.6f} ± {:.6f}".format(nu_mean, nu_se))
    print("zeta = {:.6f} ± {:.6f}  (parabola approx)".format(z_fix, zeta_se))
    print("S*   = {:.6f}  (baseline min S)".format(base['S']))

    return dict(label=label, pc=pc_mean, dpc=pc_se, nu=nu_mean, dnu=nu_se,
                zeta=z_fix, dzeta=zeta_se, S=base['S'])

# ===== MAIN =====
all_results = []
for path, label in DATASETS:
    all_results.append(run_one_dataset(path, label))

print("\n==================== SUMMARY (side-by-side) ====================")
print("{:>6s} | {:>10s} ± {:>9s} | {:>10s} ± {:>9s} | {:>10s} ± {:>9s} | {:>6s}".format(
    "label", "pc", "dpc", "nu", "dnu", "zeta", "dzeta", "S*"))
for r in all_results:
    print("{:>6s} | {:10.6f} ± {:9.6f} | {:10.6f} ± {:9.6f} | {:10.6f} ± {:9.6f} | {:6.4f}".format(
        r['label'], r['pc'], r['dpc'], r['nu'], r['dnu'], r['zeta'], r['dzeta'], r['S']))
print("================================================================")
```
