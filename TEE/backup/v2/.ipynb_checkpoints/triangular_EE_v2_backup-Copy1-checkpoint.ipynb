```python
from __future__ import print_function, division
import sys,os
# line 4 and line 5 below are for development purposes and ca be removed
qspin_path = os.path.join(os.getcwd(),"../../")
sys.path.insert(0,qspin_path)
from quspin.operators import hamiltonian # Hamiltonians and operators
#from quspin.basis import spinless_fermion_basis_1d # Hilbert space fermion basis
from quspin.basis import spin_basis_1d # Hilbert space spin basis
import numpy as np # generic math functions
import matplotlib.pyplot as plt # plotting library
from scipy.linalg import expm, sinm, cosm
from numpy.linalg import multi_dot
from scipy.sparse import csr_matrix, csc_matrix, coo_matrix, lil_matrix
from scipy.sparse.linalg import inv, eigs
from scipy.linalg import svdvals
from scipy.stats import unitary_group
from scipy import linalg
from numpy import linalg as LA
import pylab
import time
import numba
from numba import njit,c16,i8,u2
import sys
import math

np.set_printoptions(precision=2)
np.set_printoptions(suppress=True)
np.set_printoptions(threshold=np.inf, linewidth=np.inf)

time_start = time.time()

Lx,Ly=10,10

Lv=Lx*Ly # total # of vertex
L=3*Lv # total # of link qubits
Lp=2*Lv  # total # of plaquette

#case1= X,case2= Z, case3= ground state of TC
case=1

def idx(x,y,Lx,Ly):
    return (x % Lx) + Lx * (y % Ly)
def inv_idx(i,Lx,Ly):
    return i % Lx, i // Lx


def dir_x(x,y,Lx,Lv):
    return x+Lx*y
def dir_y(x,y,Lx,Lv):
    return x+Lx*y+Lv
def dir_xy(x,y,Lx,Lv):
    return x+Lx*y+2*Lv


def create_transformation_pv_link(Lv, Lx, Ly):
    #############################################
    ############ plaquette part #################
    #############################################
    p_indd = []
    for ip in range(Lx * Ly):
        ix, iy = ip % Lx, ip // Lx
        ipx=(ix + 1) % Lx
        ipy=(iy + 1) % Ly
        # 1つ目の三角形
        p_indd.append([
            dir_xy(ix, iy, Lx, Lv),
            dir_x(ix, ipy, Lx, Lv),
            dir_y(ix, iy, Lx, Lv)
        ])
        # 2つ目の三角形
        p_indd.append([
            dir_x(ix, iy, Lx, Lv),
            dir_y(ipx, iy, Lx, Lv),
            dir_xy(ix, iy, Lx, Lv)
        ])

    ############################################
    ############ vertex part ###################
    ############################################
    v_indd = []
    for iv in range(Lx * Ly):
        ix, iy = iv % Lx, iv // Lx
        imx=(ix - 1) % Lx
        imy=(iy - 1) % Ly
        v_indd.append([
            dir_x(ix, iy, Lx, Lv),
            dir_xy(ix, iy, Lx, Lv),
            dir_y(ix, iy, Lx, Lv),
            dir_x(imx, iy, Lx, Lv),
            dir_xy(imx, imy, Lx, Lv),
            dir_y(ix, imy, Lx, Lv)
        ])
    return p_indd, v_indd
"""
v_e_indd = np.zeros((Lx*Ly,6),dtype='uint32')
for iy in range(Ly):
    for ix in range(Lx):
        ic=(ix%Lx)+Lx*(iy%Ly)
        imx=(ix-1)%Lx
        imy=(iy-1)%Ly
        v_e_indd[ic][0]=dir_x(ix,iy,Lx,Lv)
        v_e_indd[ic][1]=dir_xy(ix,iy,Lx,Lv)
        v_e_indd[ic][2]=dir_y(ix,iy,Lx,Lv)
        v_e_indd[ic][3]=dir_x(imx,iy,Lx,Lv)
        v_e_indd[ic][4]=dir_xy(imx,imy,Lx,Lv)
        v_e_indd[ic][5]=dir_y(ix,imy,Lx,Lv)
#######################################
# Create plaquette operator edge-labels
p_e_indd = np.zeros((2*Lx*Ly,3),dtype='uint32')
for iy in range(Ly):
    for ix in range(Lx):
        #下三角
        ic=(ix%Lx)+Lx*(iy%Ly)
        ipx=(ix+1)%Lx
        ipy=(iy+1)%Ly
        p_e_indd[ic][0]=dir_x(ix,iy,Lx,Lv)
        p_e_indd[ic][1]=dir_y(ipx,iy,Lx,Lv)
        p_e_indd[ic][2]=dir_xy(ix,iy,Lx,Lv)
        #上三角
        p_e_indd[ic+Lx*Ly][0]=dir_xy(ix,iy,Lx,Lv)
        p_e_indd[ic+Lx*Ly][1]=dir_x(ix,ipy,Lx,Lv)
        p_e_indd[ic+Lx*Ly][2]=dir_y(ix,iy,Lx,Lv)

#print(v_e_indd)
#print(p_e_indd)
"""
####################################################
def initial_stabilizer_state(case,Ld,Lxd,Lyd,v_ind,p_ind):
    # Ld: total # of qubits
    MRi=np.zeros((Ld,2*Ld),dtype='uint32')
    if case == 1:
        #Case I: X product state
        for ii in range(Ld):
            MRi[ii][ii]=1 # X
    if case == 2:
        #Case I: Z product state
        for ii in range(Ld):
            MRi[ii][Ld+ii]=1 # Z
    if case ==3:
        ##Case III: unique g.s. of TC 
        ### plaquette stabilizers ZZZ ###
        for ip in range(2*Lxd*Lyd-1):
            MRi[ip][Ld+p_ind[ip][0]]=1 # Z0
            MRi[ip][Ld+p_ind[ip][1]]=1 # Z1
            MRi[ip][Ld+p_ind[ip][2]]=1 # Z2
            #独立ではない最後のBpだけ取らない
        ### vertex stabilizers  XXXXXX ###
        for iv in range((Lxd)*(Lyd)-1):
            ivs=iv+2*Lxd*Lyd-1  # note that iv and ivs are different.
            MRi[ivs][v_ind[iv][0]]=1 # X0
            MRi[ivs][v_ind[iv][1]]=1 # X1
            MRi[ivs][v_ind[iv][2]]=1 # X2
            MRi[ivs][v_ind[iv][3]]=1 # X3
            MRi[ivs][v_ind[iv][4]]=1 # X4
            MRi[ivs][v_ind[iv][5]]=1 # X5
            #独立ではない最後のAvだけ取らない
        ### Add logical operators###
            # x-direction loop logical Z
            for kk in range(Lx):
                iv=(kk%Lx)+Lx*(0%Ly)
                MRi[Ld-2][Ld+v_ind[iv][0]]=0 #Z
            # x-direction loop logical Z
            for kk in range(Ly):
                iv=(0%Lx)+Lx*(kk%Ly)
                MRi[Ld-1][Ld+v_ind[iv][1]]=0 #Z
    #print(MRi)
    return MRi

p_indd, v_indd=create_transformation_pv_link(Lv, Lx, Ly)

MR0=initial_stabilizer_state(case,L,Lx,Ly,v_e_indd,p_e_indd)
#print(MR0)


####################################################
def measurement_op(dMR,Ld,Lxd,Lyd,v_ind,p_ind,Gcd,mtyped):
    Meo=np.zeros(2*Ld,dtype='uint32')
    #
    if mtyped==1: #plaquette or star operator
        todo=[0,1]
        prob_list = [0.5,0.5]
        vorp = np.random.choice(todo,size=None,replace=True, p=prob_list)
        if vorp==0:
            todo=np.arange(Lxd*Lyd)
            vp=np.random.choice(todo,size=None,replace=True,p=None)
            #please check
            Meo[v_ind[vp][0]]=1 #X0
            Meo[v_ind[vp][1]]=1 #X1
            Meo[v_ind[vp][2]]=1 #X2
            Meo[v_ind[vp][3]]=1 #X3
            Meo[v_ind[vp][4]]=1 #X4
            Meo[v_ind[vp][5]]=1 #X5
        else:
            todo=np.arange(2*Lxd*Lyd)
            pp=np.random.choice(todo,size=None,replace=True,p=None)
            #please check
            Meo[Ld+p_ind[pp][0]]=1 #Z0
            Meo[Ld+p_ind[pp][1]]=1 #Z1
            Meo[Ld+p_ind[pp][2]]=1 #Z2
            
    if mtyped==2: #local X
        todo=np.arange(Ld)
        mp=np.random.choice(todo,size=None,replace=True,p=None)
        #please check
        Meo[mp]=1 #X0
        
    if mtyped==3: #local Z
        todo=np.arange(Ld)
        mp=np.random.choice(todo,size=None,replace=True,p=None)
        #please check
        Meo[Ld+mp]=1 #X0
    
    # Meo without outcome sign
    MeoF=np.zeros(2*Ld,dtype='uint32')
    MeoF=Meo
    
    # dMR without foctor
    dMRF=np.zeros((Ld,2*Ld),dtype='uint32')
    dMRF=dMR
    #check anti commtation
    Mgs=np.dot(dMRF,np.dot(Gcd,MeoF.T))
    Mgs=Mgs%2  ## check: reduce Z2 value 
    
    antic_index_list=[]
    aid=np.where(Mgs[:]!=0)
    antic_index_list=aid[0]
    #######
    lenMe=len(antic_index_list)
    if lenMe !=0:
        kc=antic_index_list[0]
        dMR_prev=dMR[kc].copy()
        #replace stabilizer for kc
        dMR[kc]=Meo
        update_list=np.delete(antic_index_list,0)
        dMR[update_list,:]=np.mod(dMR[update_list,:]+dMR_prev,2)
    return dMR


### check matrix ###
Gc=np.zeros((2*L,2*L),dtype='uint32') # 2L * 2L matrix
for k in range(L):
    Gc[k][L+k]=1
    Gc[L+k][k]=1  
Gc_csr=csr_matrix(Gc)

def subset2(Ld,Ax,Ay,Lx,Ly):
    i = idx(Ax,Ay,Lx,Ly)
    Asub_set = set()
    for j in range(3):
        Asub_set.add(p_e_indd[i][j]) #三角形
        Asub_set.add(p_e_indd[i + Lx*Ly][j]) #逆三角形
        print(p_e_indd[i][j])
        print(p_e_indd[i + Lx*Ly][j])
    return Asub_set

def subset6(Ld,Ax,Ay,Lx,Ly):
    Asub_set = set()
    Amx=(Ax - 1) % Lx
    Amy=(Ay - 1) % Ly
    i0 = idx(Ax,Ay,Lx,Ly)
    i1 = idx(Amx,Amy,Lx,Ly)
    i2 = idx(Amx,Ay,Lx,Ly)
    i3 = idx(Ax,Amy,Lx,Ly)
    for j in range(3):
        Asub_set.add(p_e_indd[i0][j])
        Asub_set.add(p_e_indd[i0 + Lx*Ly][j])
        Asub_set.add(p_e_indd[i1][j])
        Asub_set.add(p_e_indd[i1 + Lx*Ly][j])
        Asub_set.add(p_e_indd[i2][j])
        Asub_set.add(p_e_indd[i3+Lx*Ly][j])
        #print(p_e_indd[i0][j])
        #print(p_e_indd[i0 + Lx*Ly][j])
        #print(p_e_indd[i1][j])
        #print(p_e_indd[i1 + Lx*Ly][j])
        #print(p_e_indd[i0][j])
        #print(p_e_indd[i2][j])
        ##print(p_e_indd[i3 + Lx*Ly][j])
    #print("Asubset:", Asub_set)
    return Asub_set

# https://arxiv.org/abs/2204.08489
def EE_cal(MRd,Asub_set,Ld):
    # partition #
    #LA=Ld//2
    Asub=list(Asub_set)
    LA=len(Asub)
    MRdA=np.zeros((Ld,2*(len(Asub))),dtype='uint32')
    #print("new",Asub)
    for k2 in range(len(Asub)):
        MRdA[:,k2]=MRd[:,Asub[k2]].copy()
        MRdA[:,(LA)+k2]=MRd[:,(Ld)+Asub[k2]].copy()
    #print(MRdA.shape)
    #print(MRdA)
    rankA=rank_mod2_v3(MRdA)
    #print(rankA)
    EEd=rankA-(LA)
    return EEd

#############################
def rank_mod2_v3(MRdd):
    # MRdA construct
    i = 0
    Ic = MRdd.shape[0]
    Jc = MRdd.shape[1]
    for j in range(Jc):
        check_ind=np.where(MRdd[i:,j]==1)
        if len(check_ind[0]) !=0:
            if len(check_ind[0]) >1:
                elim_ind=np.delete(check_ind,0)
                MRdd[elim_ind+i,:]=np.mod(MRdd[elim_ind+i,:]+MRdd[check_ind[0][0]+i,:],2)
            MRdd[i,:],MRdd[check_ind[0][0]+i,:] = MRdd[check_ind[0][0]+i,:],MRdd[i,:].copy()
            i=i+1
    return i
############################

MR=initial_stabilizer_state(3,L,Lx,Ly,v_indd,p_indd)
#print(MR)

MR0=initial_stabilizer_state(case,L,Lx,Ly,v_e_indd,p_e_indd)#


Np=3
for ip in range(1,Np,1):
    Nd = 10 #800 sample number
    #for ip in range(len(p_list)):
    px=0.0
    pg=(1-px)*ip/(Np-1)
    pz=1-px-pg
    # stabilizer matrix
    MR=np.zeros((L,2*L),dtype='uint32') # L * 2L matrix
    # sample set of physical quantity
    MR0=initial_stabilizer_state(case,L,Lx,Ly,v_e_indd,p_e_indd)#
    TEE_list=[]
    for ids in range(Nd):
        MR=MR0
        #measurement dynamics
        NT=4*L #*L #4*L total time step
        prob_list=[pg,px,pz]
        todo=[1,2,3]
        for it in range(NT):
            mtype= np.random.choice(todo,size=None,replace=True,p=prob_list)
            MR=measurement_op(MR,L,Lx,Ly,v_e_indd,p_e_indd,Gc,mtype)
            if it==NT-1:
                Asub_set = set()
                Asub_set = subset6(L,5,5,Lx,Ly)
                SA=EE_cal(MR,Asub_set,L)
                print("Asub_set",Asub_set)
                #print(5,5,SA)
                
                Bsub_set = set()
                Bsub_set = subset6(L,3,4,Lx,Ly)
                SB=EE_cal(MR,Bsub_set,L)
                #print("Bsub_set",Bsub_set)
                #print(3,4,SB)
                
                Bsub_set=Bsub_set-Asub_set
                SB=EE_cal(MR,Bsub_set,L)
                print("Bsub_set",Bsub_set)
                #print(3,4,SB)
                
                
                Csub_set = set()
                Csub_set = subset6(L,4,3,Lx,Ly)
                SC=EE_cal(MR,Csub_set,L)
                #print("Csub_set",Csub_set)
                #print(4,3,SC)
                
                Csub_set=Csub_set-Asub_set-Bsub_set
                SC=EE_cal(MR,Csub_set,L)
                print("Csub_set",Csub_set)
                #print(4,3,SC)
                
                ABsub_set = set()
                ABsub_set = Asub_set | Bsub_set
                SAB=EE_cal(MR,ABsub_set,L)
                #print("SAB=",SAB)
                
                BCsub_set = set()
                BCsub_set = Bsub_set | Csub_set
                SBC=EE_cal(MR,BCsub_set,L)
                #print("SBC=",SBC)
                
                CAsub_set = set()
                CAsub_set = Csub_set | Asub_set
                SCA=EE_cal(MR,CAsub_set,L)
                #print("SCA=",SCA)
                
                ABCsub_set = set()
                ABCsub_set = ABsub_set | Csub_set
                SABC=EE_cal(MR,ABCsub_set,L)
                #print("SABC=",SABC)
    
                TEE=SA+SB+SC-SAB-SBC-SCA+SABC
                TEE_list=np.append(TEE_list,TEE)
            TEE_aved=np.mean(TEE_list)
            TEE_vard=np.var(TEE_list,axis=0)
            TEE_error=np.std(TEE_list,axis=0, ddof=1)/np.sqrt(len(TEE_list))
    print("SA=",SA,"SB=",SB,"SC=",SC,"SAB=",SAB,"SBC=",SBC,"SCA=",SCA,"SABC=",SABC)
    print("pg=",pg,"TEE_aved=",TEE_aved,"TEE_error=",TEE_error,"TEE_vard=",TEE_vard)
    time_end = time.time()
    print("time=",int(time_end-time_start))
    with open("Lx="+str(Lx)+"_Ly="+str(Ly)+"_L="+str(L)+"_T="+str(NT)+".lst", 'a') as f:
        f.writelines(str(pg)+'\t'+str(TEE_aved)+'\t'+str(TEE_error)+'\t'+str(TEE_vard)+"\n")

#print("pg=",pg,"SAd=",SA_aved)
```

```python
from __future__ import print_function, division
import sys,os
# line 4 and line 5 below are for development purposes and ca be removed
qspin_path = os.path.join(os.getcwd(),"../../")
sys.path.insert(0,qspin_path)
from quspin.operators import hamiltonian # Hamiltonians and operators
#from quspin.basis import spinless_fermion_basis_1d # Hilbert space fermion basis
from quspin.basis import spin_basis_1d # Hilbert space spin basis
import numpy as np # generic math functions
import matplotlib.pyplot as plt # plotting library
from scipy.linalg import expm, sinm, cosm
from numpy.linalg import multi_dot
from scipy.sparse import csr_matrix, csc_matrix, coo_matrix, lil_matrix
from scipy.sparse.linalg import inv, eigs
from scipy.linalg import svdvals
from scipy.stats import unitary_group
from scipy import linalg
from numpy import linalg as LA
import pylab
import time
import numba
from numba import njit,c16,i8,u2
import sys
import math

np.set_printoptions(precision=2)
np.set_printoptions(suppress=True)
np.set_printoptions(threshold=np.inf, linewidth=np.inf)

time_sta = time.time()

####################################################
def initial_stabilizer_state(case,Ld,Lxd,Lyd,v_ind,p_ind):
    # Ld: total # of qubits
    MRi=np.zeros((Ld,2*Ld),dtype='uint32')
    if case == 1:
        #Case I: X product state
        for ii in range(Ld):
            MRi[ii][ii]=1 # X
    if case == 2:
        #Case I: Z product state
        for ii in range(Ld):
            MRi[ii][Ld+ii]=1 # Z
    if case ==3:
        ##Case III: unique g.s. of TC 
        ### plaquette stabilizers ZZZ ###
        for ip in range(2*Lxd*Lyd-1):
            MRi[ip][Ld+p_ind[ip][0]]=1 # Z0
            MRi[ip][Ld+p_ind[ip][1]]=1 # Z1
            MRi[ip][Ld+p_ind[ip][2]]=1 # Z2
            #独立ではない最後のBpだけ取らない
        ### vertex stabilizers  XXXXXX ###
        for iv in range((Lxd)*(Lyd)-1):
            ivs=iv+2*Lxd*Lyd-1  # note that iv and ivs are different.
            MRi[ivs][v_ind[iv][0]]=1 # X0
            MRi[ivs][v_ind[iv][1]]=1 # X1
            MRi[ivs][v_ind[iv][2]]=1 # X2
            MRi[ivs][v_ind[iv][3]]=1 # X3
            MRi[ivs][v_ind[iv][4]]=1 # X4
            MRi[ivs][v_ind[iv][5]]=1 # X5
            #独立ではない最後のAvだけ取らない
        ### Add logical operators###
            # x-direction loop logical Z
            for kk in range(Lx):
                iv=(kk%Lx)+Lx*(0%Ly)
                MRi[Ld-2][Ld+v_ind[iv][0]]=0 #Z
            # x-direction loop logical Z
            for kk in range(Ly):
                iv=(0%Lx)+Lx*(kk%Ly)
                MRi[Ld-1][Ld+v_ind[iv][1]]=1 #Z
    #print(MRi)
    return MRi


####################################################
def measurement_op(dMR,Ld,Lxd,Lyd,v_ind,p_ind,Gcd,mtyped):
    Meo=np.zeros(2*Ld,dtype='uint32')
    #
    if mtyped==1: #plaquette or star operator
        todo=[0,1]
        prob_list = [0.5,0.5]
        vorp = np.random.choice(todo,size=None,replace=True, p=prob_list)
        if vorp==0:
            todo=np.arange(Lxd*Lyd)
            vp=np.random.choice(todo,size=None,replace=True,p=None)
            #please check
            Meo[v_ind[vp][0]]=1 #X0
            Meo[v_ind[vp][1]]=1 #X1
            Meo[v_ind[vp][2]]=1 #X2
            Meo[v_ind[vp][3]]=1 #X3
            Meo[v_ind[vp][4]]=1 #X4
            Meo[v_ind[vp][5]]=1 #X5
        else:
            todo=np.arange(2*Lxd*Lyd)
            pp=np.random.choice(todo,size=None,replace=True,p=None)
            #please check
            Meo[Ld+p_ind[pp][0]]=1 #Z0
            Meo[Ld+p_ind[pp][1]]=1 #Z1
            Meo[Ld+p_ind[pp][2]]=1 #Z2
            
    if mtyped==2: #local X
        todo=np.arange(Ld)
        mp=np.random.choice(todo,size=None,replace=True,p=None)
        #please check
        Meo[mp]=1 #X0
        
    if mtyped==3: #local Z
        todo=np.arange(Ld)
        mp=np.random.choice(todo,size=None,replace=True,p=None)
        #please check
        Meo[Ld+mp]=1 #X0
    
    # Meo without outcome sign
    MeoF=np.zeros(2*Ld,dtype='uint32')
    MeoF=Meo
    
    # dMR without foctor
    dMRF=np.zeros((Ld,2*Ld),dtype='uint32')
    dMRF=dMR
    #check anti commtation
    Mgs=np.dot(dMRF,np.dot(Gcd,MeoF.T))
    Mgs=Mgs%2  ## check: reduce Z2 value 
    
    antic_index_list=[]
    aid=np.where(Mgs[:]!=0)
    antic_index_list=aid[0]
    #######
    lenMe=len(antic_index_list)
    if lenMe !=0:
        kc=antic_index_list[0]
        dMR_prev=dMR[kc].copy()
        #replace stabilizer for kc
        dMR[kc]=Meo
        update_list=np.delete(antic_index_list,0)
        dMR[update_list,:]=np.mod(dMR[update_list,:]+dMR_prev,2)
    return dMR

####################
# Asub_set Asub_setを合わせていく
####################
def subset(Ld,Ax, Ay):
    i = idx(Ax, Ay)
    Asub_set = set()
    for j in range(3):
        Asub_set.add(p_e_indd[i][j])
        Asub_set.add(p_e_indd[i + Ld // 3][j])
    return Asub_set

####################
#EE
####################
# https://arxiv.org/abs/2204.08489
def EE_cal(MRd,Asub_set,Ld,p_ind):
    # partition #
    #LA=Ld//2
    Asub=list(Asub_set)
    LA=len(Asub)
    MRdA=np.zeros((Ld,2*(len(Asub))),dtype='uint32')
    #print("new",Asub)
    for k2 in range(len(Asub)):
        MRdA[:,k2]=MRd[:,Asub[k2]].copy()
        MRdA[:,(LA)+k2]=MRd[:,(Ld)+Asub[k2]].copy()
    #print(MRdA.shape)
    #print(MRdA)
    rankA=rank_mod2_v3(MRdA)
    #print(rankA)
    EEd=rankA-(LA)
    return EEd

#############################
def rank_mod2_v3(MRdd):
    # MRdA construct
    i = 0
    Ic = MRdd.shape[0]
    Jc = MRdd.shape[1]
    for j in range(Jc):
        check_ind=np.where(MRdd[i:,j]==1)
        if len(check_ind[0]) !=0:
            if len(check_ind[0]) >1:
                elim_ind=np.delete(check_ind,0)
                MRdd[elim_ind+i,:]=np.mod(MRdd[elim_ind+i,:]+MRdd[check_ind[0][0]+i,:],2)
            MRdd[i,:],MRdd[check_ind[0][0]+i,:] = MRdd[check_ind[0][0]+i,:],MRdd[i,:].copy()
            i=i+1
    return i
############################

### Main part ###
## triangular lattice under PBC##
## parameter
#set the vertex lattice system size 
Lx=3
Ly=4

Lv=Lx*Ly # total # of vertex
L=3*Lv # total # of link qubits
Lp=2*Lv  # total # of plaquette

case=1 # case=1:X product state, case=2:Z product state, case=3:unique g.s. of TC  

###############################################
# Create 3-directed edge index for each vertex
###############################################
#v_3e_indd = np.zeros((Lx*Ly,3),dtype='uint32')
def idx(x, y):
    return (x % Lx)+Lx*(y % Ly)
def inv_idx(i):
    return i % Lx, i // Lx
# 方向 0: (x+1, y), 1: (x, y+1), 2: (x-1, y+1)
directions = [(+1, 0), (0, +1), (-1, +1)]
# 一意な辺のリスト（方向0,1,2のみ使って構築）
edge_set = set()
for y in range(Ly):
    for x in range(Lx):
        i = idx(x, y)
        for dx, dy in directions:
            j = idx(x + dx, y + dy)
            edge = tuple(sorted((i, j)))
            edge_set.add(edge)
# 辞書順に並べてedge indexを定義
sorted_edges = sorted(edge_set)
edge_to_index = {e: n for n, e in enumerate(sorted_edges)}



def create_transformation_pv_link(Lv, Lx, Ly):
    #############################################
    ############ plaquette part #################
    #############################################
    p_indd = []
    for ip in range(Lx * Ly):
        ix, iy = ip % Lx, ip // Lx
        ipx=(ix + 1) % Lx
        ipy=(iy + 1) % Ly
        # 1つ目の三角形
        p_indd.append([
            dir_xy(ix, iy, Lx, Lv),
            dir_x(ix, ipy, Lx, Lv),
            dir_y(ix, iy, Lx, Lv)
        ])
        # 2つ目の三角形
        p_indd.append([
            dir_x(ix, iy, Lx, Lv),
            dir_y(ipx, iy, Lx, Lv),
            dir_xy(ix, iy, Lx, Lv)
        ])

    ############################################
    ############ vertex part ###################
    ############################################
    v_indd = []
    for iv in range(Lx * Ly):
        ix, iy = iv % Lx, iv // Lx
        imx=(ix - 1) % Lx
        imy=(iy - 1) % Ly
        v_indd.append([
            dir_x(ix, iy, Lx, Lv),
            dir_xy(ix, iy, Lx, Lv),
            dir_y(ix, iy, Lx, Lv),
            dir_x(imx, iy, Lx, Lv),
            dir_xy(imx, imy, Lx, Lv),
            dir_y(ix, imy, Lx, Lv)
        ])
    return p_indd, v_indd



####################################
# 関数：vertex index i に対して方向 0, 1, 2 の edge_index を返す
def get_3edge_indices_from_vertex(i):
    x, y = inv_idx(i)
    indices = []
    for dx, dy in directions:
        j = idx(x + dx, y + dy)
        edge = tuple(sorted((i, j)))
        indices.append(edge_to_index[edge])
    return indices
####################################
# Create vertex operator edge-labels
v_e_indd = np.zeros((Lx*Ly,6),dtype='uint32')
for iy in range(Ly):
    for ix in range(Lx):
        ic=(ix%Lx)+Lx*(iy%Ly)
        icmx=((ix-1)%Lx)+Lx*(iy%Ly)
        icmy=((ix)%Lx)+Lx*((iy-1)%Ly)
        icpxmy=((ix+1)%Lx)+Lx*((iy-1)%Ly)       
        v_e_indd[ic][0]=get_3edge_indices_from_vertex(ic)[0]
        v_e_indd[ic][1]=get_3edge_indices_from_vertex(ic)[1]
        v_e_indd[ic][2]=get_3edge_indices_from_vertex(ic)[2]
        v_e_indd[ic][3]=get_3edge_indices_from_vertex(icmx)[0]
        v_e_indd[ic][4]=get_3edge_indices_from_vertex(icmy)[1]
        v_e_indd[ic][5]=get_3edge_indices_from_vertex(icpxmy)[2]
#######################################
# Create plaquette operator edge-labels
p_e_indd = np.zeros((2*Lx*Ly,3),dtype='uint32')
for iy in range(Ly):
    for ix in range(Lx):
        #上三角
        ic=(ix%Lx)+Lx*(iy%Ly)
        icpx=((ix+1)%Lx)+Lx*(iy%Ly)
        p_e_indd[ic][0]=get_3edge_indices_from_vertex(ic)[0]
        p_e_indd[ic][1]=get_3edge_indices_from_vertex(icpx)[2]
        p_e_indd[ic][2]=get_3edge_indices_from_vertex(ic)[1]
        #下三角
        ic=(ix%Lx)+Lx*(iy%Ly)
        icpy=((ix)%Lx)+Lx*((iy+1)%Ly)
        p_e_indd[ic+Lx*Ly][0]=get_3edge_indices_from_vertex(icpx)[1]
        p_e_indd[ic+Lx*Ly][1]=get_3edge_indices_from_vertex(icpy)[0]
        p_e_indd[ic+Lx*Ly][2]=get_3edge_indices_from_vertex(icpx)[2]
### check matrix ###
Gc=np.zeros((2*L,2*L),dtype='uint32') # 2L * 2L matrix
for k in range(L):
    Gc[k][L+k]=1
    Gc[L+k][k]=1  
Gc_csr=csr_matrix(Gc)

### Main part ###
Asub_set = set()
for Ay in range(Ly-1):
    for Ax in range(Lx-1):
        Np=11
        print("Ax,Ay=",Ax,Ay)
        for ip in range(Np):
            Nd = 800 #800 sample number
            #for ip in range(len(p_list)):
            px=0.0
            pg=(1-px)*ip/(Np-1)
            pz=1-px-pg
            print("pg=",pg)
            # stabilizer matrix
            MR=np.zeros((L,2*L),dtype='uint32') # L * 2L matrix
            # sample set of physical quantity
            MR0=initial_stabilizer_state(case,L,Lx,Ly,v_e_indd,p_e_indd)#
            SA_list=[]
            for ids in range(Nd):
                MR=MR0
                #measurement dynamics
                NT=4*L #4*L total time step
                prob_list=[pg,px,pz]
                todo=[1,2,3]
                for it in range(NT):
                    mtype= np.random.choice(todo,size=None,replace=True,p=prob_list)
                    MR=measurement_op(MR,L,Lx,Ly,v_e_indd,p_e_indd,Gc,mtype)
                    if it==NT-1:
                        Asub_set |= subset(L,Ax, Ay)  # ← Asub_setのedgeの和集合
                        #print("累積された Asubset:", Asub_set)
                        SAd=EE_cal(MR,Asub_set,L,p_e_indd)
                        #print(MR.shape)
                        #print(MR)
                        SA_list=np.append(SA_list,SAd)
            print("Asubset:", Asub_set)
            SA_aved=np.mean(SA_list)
            SA_vard=np.var(SA_list,axis=0)
            SA_error=np.std(SA_list,axis=0, ddof=1)/np.sqrt(len(SA_list))
            if Ay ==0:
                with open("Ax="+str(Ax)+"_Ay="+str(Ay)+"_P="+str(2*(Ax+1)+2)+"_L="+str(L)+"_T="+str(NT)+".lst", 'a') as f:
                    f.writelines(str(pg)+'\t'+str(SA_aved)+'\t'+str(SA_vard)+'\t'+str(SA_error)+"\n")
                print("P=",2*(Ax+1)+2,"S_A=",SA_aved)
            if Ay >0:
                with open("Ax="+str(Ax)+"_Ay="+str(Ay)+"_P="+str(2*(Lx-1)+2*(Ay+1))+"_L="+str(L)+"_T="+str(NT)+".lst", 'a') as f:
                    f.writelines(str(pg)+'\t'+str(SA_aved)+'\t'+str(SA_vard)+'\t'+str(SA_error)+"\n")
                print("P=",2*(Lx+1)+2,"S_A=",SA_aved)
```
