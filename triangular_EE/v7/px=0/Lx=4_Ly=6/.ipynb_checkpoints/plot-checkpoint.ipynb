```python
import os
import re

# データを抽出するフォルダのパス
folder_path = os.getcwd() 

# 抽出結果を格納するリスト
extracted_data = []

# すべてのファイルに対して処理を行う
for filename in os.listdir(folder_path):
    if filename.endswith('.lst'):  # 対象とするファイル拡張子
        filepath = os.path.join(folder_path, filename)
        with open(filepath, 'r', encoding='utf-8') as f:
            lines = f.readlines()
            
            # 例：2行目（インデックス1）を取り出す
            if len(lines) > 1:
                extracted_data.append(lines[1].strip())
                
for line in extracted_data:
    print(line.replace('\t', ','))

```

```python
import os
import re

folder_path = os.getcwd() 
extracted_data = []

for i in range:
    for filename in os.listdir(folder_path):
        if filename.endswith('.lst'):
            filepath = os.path.join(folder_path, filename)
            with open(filepath, 'r', encoding='utf-8') as f:
                lines = f.readlines()
                data_line = lines[i].strip()
                base_name = os.path.splitext(filename)[0]
                match = re.search(r'P=(\d+)', base_name)
                if match:
                    param = match.group(1)
                    combined_line = param + '\t' + data_line
                    extracted_data.append(combined_line)
    
    sorted_data = sorted(extracted_data, key=lambda x: int(x.split('\t')[0]))
    
    for line in sorted_data:
        print(line)
    
    # sorted_data の1行目を例に取り出す
    first_line = sorted_data[0]
    # タブ区切りで分割し、2列目を取り出す（インデックス1）
    value_for_filename = first_line.split('\t')[1]
    
    # 小数点を含むのでファイル名に使うなら安全な文字列に変換（例：小数点をアンダースコアに）
    safe_value = value_for_filename.replace('.', '_')
    
    output_file = f'output_{safe_value}.txt'
    
    with open(output_file, 'w', encoding='utf-8') as f_out:
        for line in sorted_data:
            f_out.write(line + '\n')
```

```python
import os
import re

folder_path = os.getcwd() 

# 例えば10行目まで処理したい場合
extracted_data = []  # iごとにリセット
for i in range(10):  

    for filename in os.listdir(folder_path):
        if filename.endswith('.lst'):
            filepath = os.path.join(folder_path, filename)
            with open(filepath, 'r', encoding='utf-8') as f:
                lines = f.readlines()
                if len(lines) > i:
                    data_line = lines[i].strip()
                    base_name = os.path.splitext(filename)[0]
                    match = re.search(r'P=(\d+)', base_name)
                    if match:
                        param = match.group(1)
                        combined_line = param + '\t' + data_line
                        extracted_data.append(combined_line)

    sorted_data = sorted(extracted_data, key=lambda x: int(x.split('\t')[0]))

    for line in sorted_data:
        print(line)

    # ファイル名用の値取得と変換
    if sorted_data:
        first_line = sorted_data[0]
        value_for_filename = first_line.split('\t')[1]
        safe_value = value_for_filename.replace('.', '_')
        output_file = f'pg={safe_value}.txt'

        with open(output_file, 'w', encoding='utf-8') as f_out:
            for line in sorted_data:
                f_out.write(line + '\n')
```

```python
import matplotlib.pyplot as plt

# 与えられたデータ（タブ区切りの文字列をリスト化）
data_lines = [
    "4\t0.3\t0.39125\t0.37817343750000004\t0.021755652884147592",
    "6\t0.3\t0.61375\t0.5820609375\t0.026990494259098666",
    "8\t0.3\t0.83\t0.7186000000000001\t0.02998956848259433",
    "8\t0.3\t0.6925\t0.6254437500000001\t0.02797826239168674",
    "10\t0.3\t0.88125\t0.8546484375\t0.032705467530414215",
    "10\t0.3\t1.09\t1.0219\t0.035762742024996255"
]

# 1列目（x軸）, 3列目（y軸）, 5列目（誤差）を抽出
x = []
y = []
yerr = []

for line in data_lines:
    cols = line.split('\t')
    x.append(int(cols[0]))
    y.append(float(cols[2]))
    yerr.append(float(cols[4]))

# グラフ描画
plt.errorbar(x, y, yerr=yerr, fmt='o', capsize=5, linestyle='-', color='blue')

plt.xlabel('P')
plt.ylabel('S_A')
plt.title('test')
plt.grid(True)
plt.show()
```
