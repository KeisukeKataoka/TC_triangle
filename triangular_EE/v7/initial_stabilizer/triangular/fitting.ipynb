```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# パラメータ設定
Lx, Ly = 400, 40000
x1 = np.arange(4, 2*Lx + 1, 2)              # x = 4, 6, ..., 20
y1 = x1 - 1                                 # y = x - 1

x2 = np.arange(2*Lx + 2, 2*Lx + 2*Ly + 1, 2) # x = 22, 24, ..., 30
y2 = x2 - 2                              # y = x - 1.95

# 全体のデータ結合
x_all = np.concatenate([x1, x2])
y_all = np.concatenate([y1, y2])

# 線形関数定義
def linear(x, a, b):
    return a * x + b

# 最小二乗フィッティング
popt, _ = curve_fit(linear, x_all, y_all)
a, b = popt

# 結果表示
print(f" y = {a:.5f} x + {b:.5f}")

# グラフ描画
x_fit = np.linspace(min(x_all), max(x_all), 300)
y_fit = linear(x_fit, *popt)

plt.plot(x_all, y_all, 'o', label='data', color='black')
plt.plot(x_fit, y_fit, 'r-', label=f' y = {a:.3f}x + {b:.3f}')
plt.xlabel('x')
plt.ylabel('y')
plt.title('fitting')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
