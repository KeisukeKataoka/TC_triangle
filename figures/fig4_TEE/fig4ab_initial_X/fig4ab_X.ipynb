```python
import numpy as np
import matplotlib.pyplot as plt
import re
import os
import matplotlib as mpl

mpl.rcParams['svg.fonttype'] = 'path'  # テキストをフォントのまま保持

cmap = plt.colormaps['gnuplot']

# ファイル名を指定
filename = "Lx_24_Nd_500_NT_17280_px_0.0_MPI_hexagonal6.npz"
print("見つかったファイル:", filename)

# 情報抽出
match = re.search(r"Lx_(\d+)_Nd_(\d+)_NT_(\d+)_px_([0-9.]+)_", filename)
if match:
    Lx = int(match.group(1))
    Nd = int(match.group(2))
    NT = int(match.group(3))
    px = float(match.group(4))
    title_suffix = f"Lx={Lx}, Nd={Nd}, NT={NT}, px={px}"
else:
    title_suffix = os.path.basename(filename)

# 読み込みとプロット
with np.load(filename) as data:
    pg = data['pg']
    TEE1_ave = data['TEE1_ave']
    TEE1_err = data['TEE1_err']
    TEE1_var = data['TEE1_var']
    TEE2_ave = data['TEE2_ave']
    TEE2_err = data['TEE2_err']
    TEE2_var = data['TEE2_var']
    TEE3_ave = data['TEE3_ave']
    TEE3_err = data['TEE3_err']
    TEE3_var = data['TEE3_var']
    TEE4_ave = data['TEE4_ave']
    TEE4_err = data['TEE4_err']
    TEE4_var = data['TEE4_var']
    TEE5_ave = data['TEE5_ave']
    TEE5_err = data['TEE5_err']
    TEE5_var = data['TEE5_var']
    TEE6_ave = data['TEE6_ave']
    TEE6_err = data['TEE6_err']
    TEE6_var = data['TEE6_var']

label1 = "$A_h$=6"
label2 = "$A_h$=12"
label3 = "$A_h$=18"
label4 = "$A_h$=24"
label5 = "$A_h$=30"
label6 = "$A_h$=36"

# --- Fig4a: 平均と誤差 ---
fig4a, ax = plt.subplots(figsize=(6,5))
ax.errorbar(pg, TEE1_ave, yerr=TEE1_err, fmt='o-', capsize=5, label=label1)
ax.errorbar(pg, TEE2_ave, yerr=TEE2_err, fmt='o-', capsize=5, label=label2)
ax.errorbar(pg, TEE3_ave, yerr=TEE3_err, fmt='o-', capsize=5, label=label3)
ax.errorbar(pg, TEE4_ave, yerr=TEE4_err, fmt='o-', capsize=5, label=label4)
ax.errorbar(pg, TEE5_ave, yerr=TEE5_err, fmt='o-', capsize=5, label=label5)
ax.errorbar(pg, TEE6_ave, yerr=TEE6_err, fmt='o-', capsize=5, label=label6)

ax.set_xlabel('$p_g$', fontsize=16)
ax.set_ylabel('$S_\mathcal{T}$', fontsize=16)
ax.set_ylim(-1, 0.2)
ax.set_xlim(0.71, 0.9)
ax.tick_params(axis='both', which='major',
                direction="in", length=6, width=1.2, labelsize=12)
ax.tick_params(axis='x', which='major',pad=10)
ax.axhline(0, color='black', linewidth=1.0)
ax.legend()
plt.tight_layout()
plt.savefig("fig4ab_X/fig4a_initial_X.svg", format="svg", bbox_inches="tight", transparent=True)
plt.show()


# --- Fig4b: 分散 ---
fig4b, ax = plt.subplots(figsize=(6,5))
ax.plot(pg, TEE1_var, 's--', label=label1)
ax.plot(pg, TEE2_var, 's--', label=label2)
ax.plot(pg, TEE3_var, 's--', label=label3)
ax.plot(pg, TEE4_var, 's--', label=label4)
ax.plot(pg, TEE5_var, 's--', label=label5)
ax.plot(pg, TEE6_var, 's--', label=label6)

ax.set_xlabel('$p_g$', fontsize=16)
ax.set_ylabel('$\chi^{TE}$', fontsize=16)
ax.set_xlim(0.71, 0.9)
ax.set_ylim(0, 0.025)
ax.tick_params(axis='both', which='major',
                direction="in", length=6, width=1.2, labelsize=12)
ax.legend()
plt.tight_layout()
plt.savefig("fig4ab_X/fig4b_initial_X.svg", format="svg", bbox_inches="tight", transparent=True)
plt.show()
```
