```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
Simple percolation visualization (site or bond) on a 2D square lattice.
- Shows clusters with distinct colors
- Highlights a spanning cluster (if any)
- Prints whether vertical/horizontal spanning exists
"""

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# ======= parameters =======
L = 80            # lattice size (LxL)
p = 0.7          # occupation prob (site) or bond-open prob (bond)
KIND = "site"     # "site" or "bond"
SEED = 1          # RNG seed for reproducibility (None for random)
# ==========================

rng = np.random.default_rng(SEED)

def union_find_init(n):
    parent = np.arange(n, dtype=int)
    size = np.ones(n, dtype=int)
    return parent, size

def find(parent, x):
    while parent[x] != x:
        parent[x] = parent[parent[x]]  # path halving
        x = parent[x]
    return x

def union(parent, size, a, b):
    ra, rb = find(parent, a), find(parent, b)
    if ra == rb:
        return
    if size[ra] < size[rb]:
        ra, rb = rb, ra
    parent[rb] = ra
    size[ra] += size[rb]

def site_percolation(L, p, rng):
    # occupied sites
    occ = rng.random((L, L)) < p

    # union-find over occupied 4-neighbor connectivity (open boundaries)
    parent, size = union_find_init(L*L)
    def idx(i, j): return i*L + j

    for i in range(L):
        for j in range(L):
            if not occ[i, j]:
                continue
            # right neighbor
            if j+1 < L and occ[i, j+1]:
                union(parent, size, idx(i,j), idx(i,j+1))
            # down neighbor
            if i+1 < L and occ[i+1, j]:
                union(parent, size, idx(i,j), idx(i+1,j))

    # compute root labels only for occupied sites
    roots = -np.ones((L, L), dtype=int)
    for i in range(L):
        for j in range(L):
            if occ[i, j]:
                roots[i, j] = find(parent, idx(i, j))
    return occ, roots

def bond_percolation(L, p, rng):
    # all sites exist; bonds (horizontal/vertical) are open with prob p
    right_open = rng.random((L, L-1)) < p  # bonds between (i,j) and (i,j+1)
    down_open  = rng.random((L-1, L)) < p  # bonds between (i,j) and (i+1,j)

    parent, size = union_find_init(L*L)
    def idx(i, j): return i*L + j

    # union open bonds
    for i in range(L):
        for j in range(L-1):
            if right_open[i, j]:
                union(parent, size, idx(i,j), idx(i,j+1))
    for i in range(L-1):
        for j in range(L):
            if down_open[i, j]:
                union(parent, size, idx(i,j), idx(i+1,j))

    # label by roots (all sites present)
    roots = np.fromfunction(lambda i, j: find(parent, (i.astype(int))*L + j.astype(int)),
                            (L, L), dtype=int)
    occ = np.ones((L, L), dtype=bool)  # for compatibility
    return occ, roots

def spanning_masks(occ, roots):
    """Return masks for clusters that span vertically or horizontally."""
    L = occ.shape[0]
    # list unique root ids among occupied sites
    occ_roots = np.unique(roots[occ])
    occ_roots = occ_roots[occ_roots >= 0]

    vert_spanners = []
    hori_spanners = []
    for r in occ_roots:
        mask = (roots == r)
        # vertical: touch top (i=0) and bottom (i=L-1)
        if mask[0, :].any() and mask[L-1, :].any():
            vert_spanners.append(r)
        # horizontal: touch left (j=0) and right (j=L-1)
        if mask[:, 0].any() and mask[:, L-1].any():
            hori_spanners.append(r)

    return set(vert_spanners), set(hori_spanners)

def colorize_clusters(occ, roots):
    """Map each cluster root to a color index. -1 (unoccupied) -> 0 (background)."""
    L = occ.shape[0]
    label_map = -np.ones(L*L+1, dtype=int)  # root id -> color index
    color_id = 1  # start at 1; 0 reserved for background
    colored = np.zeros((L, L), dtype=int)

    # assign compact IDs to encountered roots
    for i in range(L):
        for j in range(L):
            r = roots[i, j]
            if r < 0:  # unoccupied for site percolation
                colored[i, j] = 0
            else:
                if label_map[r] < 0:
                    label_map[r] = color_id
                    color_id += 1
                colored[i, j] = label_map[r]
    # build a random colormap
    n_colors = max(1, color_id)
    rng_local = np.random.default_rng(1234)
    base = np.ones((n_colors, 4))
    # background = light gray
    base[0] = [0.93, 0.93, 0.93, 1.0]
    for k in range(1, n_colors):
        base[k, :3] = rng_local.random(3)  # random RGB
        base[k, 3] = 1.0
    cmap = ListedColormap(base)
    return colored, cmap

# --- run ---
if KIND.lower() == "site":
    occ, roots = site_percolation(L, p, rng)
elif KIND.lower() == "bond":
    occ, roots = bond_percolation(L, p, rng)
else:
    raise ValueError("KIND must be 'site' or 'bond'")

vspan, hspan = spanning_masks(occ, roots)
colored, cmap = colorize_clusters(occ, roots)

print(f"[{KIND}] L={L}, p={p:.3f}")
print(f" vertical spanning clusters: {len(vspan)} {'(YES)' if len(vspan)>0 else '(no)'}")
print(f" horizontal spanning clusters: {len(hspan)} {'(YES)' if len(hspan)>0 else '(no)'}")

# highlight spanning clusters with hatch overlay
L = occ.shape[0]
span_mask = np.zeros_like(occ, dtype=bool)
for r in (vspan | hspan):
    span_mask |= (roots == r)

fig, ax = plt.subplots(figsize=(6, 6))
im = ax.imshow(colored, origin="upper", interpolation="nearest", cmap=cmap)
ax.set_xticks([]); ax.set_yticks([])
ax.set_title(f"{KIND.capitalize()} percolation (L={L}, p={p:.3f})", fontsize=12)

# draw a semi-transparent overlay for spanning clusters
overlay = np.zeros((*span_mask.shape, 4))
overlay[..., :3] = 0.0  # black
overlay[..., 3] = span_mask * 0.35  # alpha only where spanning
ax.imshow(overlay, origin="upper", interpolation="nearest")

# annotate text
if len(vspan) or len(hspan):
    ax.text(0.5, -0.05, "SPANNING!", ha="center", va="top", transform=ax.transAxes,
            fontsize=12, weight="bold")
else:
    ax.text(0.5, -0.05, "no spanning", ha="center", va="top", transform=ax.transAxes,
            fontsize=10)

plt.tight_layout()
plt.show()
```

```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
import numpy as np
import matplotlib.pyplot as plt

# ======= settings =======
MODE = "bond_square"     # "bond_square" (pc=0.5) or "site_triangular" (pc=0.5)
L = 64                   # system size
TRIALS = 200             # trials per p (増やすと精度↑, ただし時間↑)
SEED = 0
PS = np.linspace(0.42, 0.58, 17)  # 掃引レンジ（必要に応じて調整）
TARGET = 0.5             # target spanning probability for binary search
BSEARCH_ITERS = 18       # 二分探索の反復回数（~2^-18 分解能）
# ========================

rng = np.random.default_rng(SEED)

# ---------- union-find ----------
def uf_init(n):
    parent = np.arange(n, dtype=int)
    size = np.ones(n, dtype=int)
    return parent, size

def uf_find(parent, x):
    # x はスカラー整数を仮定
    while parent[x] != x:
        parent[x] = parent[parent[x]]   # path halving
        x = parent[x]
    return x

def uf_union(parent, size, a, b):
    ra, rb = uf_find(parent, a), uf_find(parent, b)
    if ra == rb:
        return
    if size[ra] < size[rb]:
        ra, rb = rb, ra
    parent[rb] = ra
    size[ra] += size[rb]

def idx(i, j, L): 
    return i*L + j

# ---------- spanning check ----------
def spanning_from_roots(roots, L):
    """上下または左右に同一クラスターが接触していればスパニング True"""
    top = np.unique(roots[0, :])
    bot = np.unique(roots[L-1, :])
    lef = np.unique(roots[:, 0])
    rig = np.unique(roots[:, L-1])
    vert = np.intersect1d(top, bot)
    hori = np.intersect1d(lef, rig)
    return (vert[vert >= 0].size > 0) or (hori[hori >= 0].size > 0)

# ---------- models with pc=0.5 ----------
def bond_perc_square(L, p, rng):
    """正方格子ボンド・パーコレーション: pc=0.5 (厳密)"""
    # 水平/垂直ボンドの開閉
    right_open = rng.random((L, L-1)) < p
    down_open  = rng.random((L-1, L)) < p

    parent, size = uf_init(L*L)

    # ボンドが開いていれば union
    for i in range(L):
        for j in range(L-1):
            if right_open[i, j]:
                uf_union(parent, size, idx(i, j, L), idx(i, j+1, L))
    for i in range(L-1):
        for j in range(L):
            if down_open[i, j]:
                uf_union(parent, size, idx(i, j, L), idx(i+1, j, L))

    # roots を二重ループで作る（fromfunction を使わない）
    roots = np.empty((L, L), dtype=int)
    for i in range(L):
        for j in range(L):
            roots[i, j] = uf_find(parent, idx(i, j, L))

    return spanning_from_roots(roots, L)

def site_perc_triangular(L, p, rng):
    """三角格子サイト・パーコレーション: pc=0.5 (厳密)
       正方格子座標で 6 近傍を実現：4 近傍 + 片対角（ここでは (i+1, j-1)）"""
    occ = rng.random((L, L)) < p
    parent, size = uf_init(L*L)

    for i in range(L):
        for j in range(L):
            if not occ[i, j]:
                continue
            # 4 近傍（右・下）
            if j+1 < L and occ[i, j+1]:
                uf_union(parent, size, idx(i, j, L), idx(i, j+1, L))
            if i+1 < L and occ[i+1, j]:
                uf_union(parent, size, idx(i, j, L), idx(i+1, j, L))
            # 片対角（i+1, j-1）で三角格子化（逆向きは他セルで拾われる）
            if i+1 < L and j-1 >= 0 and occ[i+1, j-1]:
                uf_union(parent, size, idx(i, j, L), idx(i+1, j-1, L))

    # roots を二重ループで作る。未占有は -1
    roots = -np.ones((L, L), dtype=int)
    for i in range(L):
        for j in range(L):
            if occ[i, j]:
                roots[i, j] = uf_find(parent, idx(i, j, L))

    return spanning_from_roots(roots, L)

# ---------- trial runner ----------
def trial_spanning_probability(p):
    hit = 0
    if MODE == "bond_square":
        for _ in range(TRIALS):
            hit += bond_perc_square(L, p, rng)
    elif MODE == "site_triangular":
        for _ in range(TRIALS):
            hit += site_perc_triangular(L, p, rng)
    else:
        raise ValueError("MODE must be 'bond_square' or 'site_triangular'")
    return hit / TRIALS

# ---- sweep & plot ----
P = np.array([trial_spanning_probability(p) for p in PS])

plt.figure(figsize=(6, 4.5))
plt.plot(PS, P, "o-")
plt.axhline(0.5, ls="--")
plt.xlabel("p")
plt.ylabel("P_span(p)")
plt.title(f"{MODE}  L={L}, trials={TRIALS}")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ---- binary search for p*: P_span(p*) ~ TARGET ----
lo, hi = float(min(PS)), float(max(PS))
for _ in range(BSEARCH_ITERS):
    mid = 0.5 * (lo + hi)
    Pm = trial_spanning_probability(mid)
    if Pm < TARGET:
        lo = mid
    else:
        hi = mid
p_star = 0.5 * (lo + hi)
print(f"Estimated p* ≈ {p_star:.4f}  (L={L}, {MODE}, TRIALS={TRIALS})")
```
